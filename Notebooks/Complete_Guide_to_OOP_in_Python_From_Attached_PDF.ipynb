## 1. Introduction: Why Do We Need OOP?

### The Problem with Procedural Programming

Procedural programming organizes code as instructions, variables, and functions executed in sequence. This is suitable for small programs, but as an application grows—such as a banking system, game, or e-commerce application—many separate variables and functions can become difficult to manage.

### The OOP Solution

**Object-Oriented Programming (OOP)** organizes related data and behavior into objects.

- **Attributes** represent data or properties.
- **Methods** represent behavior or actions.
- **Objects** combine related attributes and methods into one unit.

### Real-world analogy: Smartphone

A smartphone has properties such as color, storage, and brand. It also has behaviors such as making calls, taking photographs, and opening applications. OOP lets us model this as a single object with related data and actions.


## 2. Classes vs Objects: Blueprint and House

### Class
A class is a blueprint or template that defines the attributes and methods objects will have.

### Object
An object is an actual instance created using a class. Multiple objects can be created from one class, and each object can hold different data.

**Analogy:** A house blueprint is the class. Individual houses constructed from that blueprint are objects. The houses may have different colors or owners, even though they follow the same blueprint.


## 3. Creating Your First Class and Object

### Key Terms

- `class`: Keyword used to define a class. Class names conventionally use PascalCase.
- `__init__()`: Initializer that runs automatically when an object is created.
- `self`: Reference to the current object; it allows the object to store its own attributes.
- **Attribute:** Data stored in an object.
- **Method:** Function defined inside a class that represents behavior.

### Example: Dog class


````python
# Define a class named Dog
class Dog:

    # Constructor/initializer: runs when a Dog object is created
    def __init__(self, name, breed, age):
        # Store the name in this particular object
        self.name = name

        # Store the breed in this particular object
        self.breed = breed

        # Store the age in this particular object
        self.age = age

    # Instance method representing a dog's behavior
    def bark(self):
        # Use the object's name when displaying the message
        print(f"{self.name} says: Woof! Woof!")

    # Instance method to increase the dog's age
    def celebrate_birthday(self):
        # Increase the age attribute by one
        self.age += 1

        # Display the updated age
        print(f"Happy Birthday {self.name}! You are now {self.age} years old.")

# Create the first Dog object
dog1 = Dog("Bruno", "Golden Retriever", 3)

# Create the second Dog object
dog2 = Dog("Bella", "Beagle", 5)

# Access attributes of dog1
print(f"{dog1.name} is a {dog1.breed}.")

# Access attributes of dog2
print(f"{dog2.name} is {dog2.age} years old.")

# Call methods using the objects
dog1.bark()
dog2.celebrate_birthday()
````

### Code Explanation

- `class Dog:` creates the class.
- `__init__()` initializes the attributes whenever a new Dog object is created.
- `self.name`, `self.breed`, and `self.age` are instance attributes.
- `bark()` is an instance method that uses the object's name.
- `celebrate_birthday()` modifies the age of the particular object.
- `dog1` and `dog2` are independent objects.

### Expected Output
```text
Bruno is a Golden Retriever.
Bella is 5 years old.
Bruno says: Woof! Woof!
Happy Birthday Bella! You are now 6 years old.
```

### Important Observation

Changing `dog2.age` does not affect `dog1.age` because each object has its own instance attributes.


## 4. Class Attributes vs Instance Attributes

### Instance Attributes
Instance attributes belong to individual objects. They are usually created inside `__init__()` using `self`.

Examples: a student's name, roll number, or marks.

### Class Attributes
Class attributes are defined directly inside the class, outside methods. They are shared by objects of the class unless an object creates its own attribute with the same name.

Example: school name shared by all students.


````python
# Define a Student class
class Student:
    # Class attribute shared by Student objects
    school_name = "Greenwood High"

    # Constructor
    def __init__(self, name, grade):
        # Instance attribute: unique to each object
        self.name = name

        # Instance attribute: unique to each object
        self.grade = grade

# Create two Student objects
s1 = Student("Ananya", 10)
s2 = Student("Rohan", 12)

# Access the shared class attribute
print(s1.school_name)
print(s2.school_name)

# Access individual instance attributes
print(s1.name, s1.grade)
print(s2.name, s2.grade)
````

### Expected Output
```text
Greenwood High
Greenwood High
Ananya 10
Rohan 12
```

### Summary

| Attribute | Defined Where | Shared? | Example |
|---|---|---|---|
| Instance attribute | Usually in `__init__()` using `self` | No; each object has its own | `self.name` |
| Class attribute | Directly inside the class | Yes, normally shared | `school_name` |

The four core principles are:

These principles help organize code, reuse functionality, protect data, and create flexible designs.


## Pillar 1: Encapsulation — Protecting Data

### Concept

Encapsulation bundles data and methods inside a class while controlling how internal data is accessed or modified.

### Analogy: ATM

An ATM provides buttons and options such as deposit, withdraw, and check balance. Customers interact with these operations without directly changing the machine's internal cash counter.

### Python Naming Convention

Prefixing an attribute with two underscores (`__`) triggers name mangling. This makes direct access from outside the class more difficult, though it is not strict security.


````python
# Define a BankAccount class
class BankAccount:

    # Constructor
    def __init__(self, owner, initial_balance):
        # Store the account owner's name
        self.owner = owner

        # Name-mangled attribute intended for internal use
        self.__balance = initial_balance

    # Method to deposit money
    def deposit(self, amount):
        # Accept only positive deposit amounts
        if amount > 0:
            # Increase the internal balance
            self.__balance += amount
            print(f"Deposited ₹{amount}.")
        else:
            # Reject invalid deposit values
            print("Invalid deposit amount!")

    # Method to withdraw money
    def withdraw(self, amount):
        # Check that the amount is positive and available
        if 0 < amount <= self.__balance:
            # Reduce the internal balance
            self.__balance -= amount
            print(f"Withdrew ₹{amount}.")
        else:
            # Handle invalid amount or insufficient balance
            print("Insufficient funds or invalid amount!")

    # Method to check the balance
    def check_balance(self):
        # Return the balance through a controlled method
        return f"Current Balance: ₹{self.__balance}"

# Create a bank account
account = BankAccount("Priya", 5000)

# Deposit money
account.deposit(1500)

# Display the current balance
print(account.check_balance())

# Attempting direct access is not recommended and normally raises AttributeError
#print(account.__balance)
````

### Expected Output
```text
Deposited ₹1500.
Current Balance: ₹6500
```

### Code Explanation

- `owner` is a public attribute.
- `__balance` is a name-mangled attribute intended to be accessed through class methods.
- `deposit()` validates and updates the balance.
- `withdraw()` checks that the withdrawal is valid and funds are available.
- `check_balance()` provides controlled access to the balance.

**Teaching note:** Python's double underscore is not a strict private-access mechanism. It performs name mangling to reduce accidental access and naming conflicts.


## Pillar 2: Inheritance — Reusing Code

### Concept

Inheritance allows a child/subclass to reuse attributes and methods of a parent/superclass.

It helps reduce repeated code and supports the **DRY (Don't Repeat Yourself)** principle.

### Analogy

A child may inherit some characteristics from parents while also developing unique traits.

### Example: Vehicle and ElectricCar

- `Vehicle` is the parent class.
- `ElectricCar` is the child class.
- `ElectricCar` inherits the `start_engine()` method.
- `super()` calls the parent class initializer.


````python
# Define the parent class
class Vehicle:

    # Parent constructor
    def __init__(self, brand, speed):
        # Store the vehicle brand
        self.brand = brand

        # Store the vehicle speed
        self.speed = speed

    # Parent method
    def start_engine(self):
        # Display an engine-start message
        print(f"{self.brand} engine started.")

# Define the child class inheriting from Vehicle
class ElectricCar(Vehicle):

    # Child constructor
    def __init__(self, brand, speed, battery_range):
        # Call the parent constructor to initialize inherited attributes
        super().__init__(brand, speed)

        # Initialize the child-specific attribute
        self.battery_range = battery_range

    # Child-specific method
    def charge(self):
        # Display charging information
        print(f"{self.brand} is charging. Max range: {self.battery_range} km.")

# Create an ElectricCar object
tesla = ElectricCar("Tesla", 200, 500)

# Call the inherited parent method
tesla.start_engine()

# Call the child-specific method
tesla.charge()
````

### Expected Output
```text
Tesla engine started.
Tesla is charging. Max range: 500 km.
```

### Code Explanation

- `class ElectricCar(Vehicle):` establishes inheritance.
- `super().__init__(brand, speed)` calls the parent initializer.
- `battery_range` is specific to ElectricCar.
- `start_engine()` is inherited from Vehicle.
- `charge()` is specific to ElectricCar.


## Pillar 3: Polymorphism — Many Forms

### Concept

Polymorphism means **many forms**. Different classes can provide methods with the same name but different behavior.

### Analogy: Speak Button

The same action can produce different behavior:
- Dog → Bark
- Cat → Meow
- Duck → Quack

Python supports polymorphism through common method names and also through method overriding in inheritance.


````python
# Define a Cat class
class Cat:
    def make_sound(self):
        # Cat-specific sound
        return "Meow!"

# Define a Cow class
class Cow:
    def make_sound(self):
        # Cow-specific sound
        return "Moo!"

# Define a Duck class
class Duck:
    def make_sound(self):
        # Duck-specific sound
        return "Quack!"

# Store objects from different classes in one list
animals = [Cat(), Cow(), Duck()]

# Call the same method on each object
for animal in animals:
    # Python uses the object's own implementation
    print(animal.make_sound())
````

### Expected Output
```text
Meow!
Moo!
Quack!
```

### Code Explanation

- Each class defines a method named `make_sound()`.
- The method implementation differs in each class.
- The loop does not need to check the object's class.
- Each object responds to the same method call in its own way.

This is an example of polymorphism through a common interface (duck typing).


### Polymorphism Through Method Overriding

Method overriding occurs when a child class defines a method with the same name as a method inherited from its parent class. The child implementation is used when called on the child object.


````python
# Define a parent class
class Animal:
    def speak(self):
        # Generic parent implementation
        print("Animal makes a sound")

# Define a child class
class Dog(Animal):
    def speak(self):
        # Override the inherited method
        print("Dog barks")

# Define another child class
class Cat(Animal):
    def speak(self):
        # Override the inherited method
        print("Cat meows")

# Create objects
dog = Dog()
cat = Cat()

# Call the overridden method
dog.speak()
cat.speak()
````

### Expected Output
```text
Dog barks
Cat meows
```


## Pillar 4: Abstraction — Hiding Complexity

### Concept

Abstraction exposes essential features while hiding implementation details. Abstract classes can also enforce a structure by requiring subclasses to implement certain methods.

### Analogy: Driving a Car

A driver uses the steering wheel, accelerator, and brake without needing to understand every internal component of the engine.

### Python ABC Module

Python provides the `abc` module:
- `ABC` is used as a base class for abstract classes.
- `@abstractmethod` marks a method that subclasses must implement.
- An abstract class cannot be instantiated while it contains unimplemented abstract methods.


````python
# Import tools for defining abstract classes
from abc import ABC, abstractmethod

# Define an abstract parent class
class PaymentGateway(ABC):

    # Declare an abstract method
    @abstractmethod
    def process_payment(self, amount):
        # Subclasses must provide implementation
        pass

# Define a UPI payment implementation
class UPIPayment(PaymentGateway):
    def process_payment(self, amount):
        # Implement payment processing for UPI
        print(f"Processing ₹{amount} via UPI QR scan...")

# Define a credit card payment implementation
class CreditCardPayment(PaymentGateway):
    def process_payment(self, amount):
        # Implement payment processing for credit card
        print(f"Processing ₹{amount} via Credit Card with OTP verification...")

# Create UPI payment object
upi = UPIPayment()

# Process a UPI payment
upi.process_payment(999)

# Create credit card payment object
card = CreditCardPayment()

# Process a card payment
card.process_payment(2500)
````

### Expected Output
```text
Processing ₹999 via UPI QR scan...
Processing ₹2500 via Credit Card with OTP verification...
```

### Code Explanation

- `PaymentGateway(ABC)` defines an abstract base class.
- `@abstractmethod` marks `process_payment()` as mandatory for subclasses.
- `UPIPayment` and `CreditCardPayment` implement the method.
- Both child classes follow the same interface but use different implementations.

If we try to create `PaymentGateway()` directly, Python raises an error because the abstract method has not been implemented.


# 6. Bonus: Special Dunder (Magic) Methods

Dunder methods are special methods whose names begin and end with double underscores. They allow objects to work with built-in Python operations.

Examples:
- `__init__()` — initializes an object.
- `__str__()` — controls the readable string representation used by `print()`.
- `__repr__()` — provides a representation intended to be useful for debugging.
- `__len__()` — supports the built-in `len()`.
- `__add__()` — custom behavior for the `+` operator.

### Example: `__str__()`


````python
# Define a Book class
class Book:

    # Constructor
    def __init__(self, title, author):
        # Store the book title
        self.title = title

        # Store the book author
        self.author = author

    # Special method controlling print(object)
    def __str__(self):
        # Return a readable string for the book
        return f"'{self.title}' by {self.author}"

# Create a Book object
my_book = Book("The Alchemist", "Paulo Coelho")

# Print the object
print(my_book)
````

### Expected Output
```text
'The Alchemist' by Paulo Coelho
```

Without `__str__()`, printing the object normally displays a default object representation rather than a friendly description.


# 7. Complete Mini-Project: Library Management System

This mini-project combines:
- Classes and objects
- Constructors
- Attributes and methods
- Encapsulation
- Object interaction
- `__str__()` method
- Conditional logic and loops

### Project Requirements

#### Book class
- Store title and author.
- Track whether the book is borrowed.
- Provide a readable string representation.

#### Library class
- Store the library name.
- Maintain a list of books.
- Add books.
- Display the catalog.
- Borrow a book by title.
- Prevent a book from being borrowed if it is already borrowed.


````python
# Define the Book class
class Book:

    # Constructor
    def __init__(self, title, author):
        # Store the title
        self.title = title

        # Store the author
        self.author = author

        # New books are initially available
        self.is_borrowed = False

    # Special method for readable book information
    def __str__(self):
        # Determine the current status
        status = "Borrowed" if self.is_borrowed else "Available"

        # Return formatted book information
        return f"{self.title} by {self.author} [{status}]"

# Define the Library class
class Library:

    # Constructor
    def __init__(self, name):
        # Store the library name
        self.name = name

        # Encapsulated list of books
        self.__books = []

    # Add a book to the library
    def add_book(self, book):
        # Add the Book object to the internal list
        self.__books.append(book)

        # Confirm that the book was added
        print(f"Added: {book.title}")

    # Display all books
    def show_books(self):
        # Print the catalog heading
        print(f"\n--- {self.name} Catalog ---")

        # Iterate through all books
        for book in self.__books:
            # Print each book using its __str__ method
            print(book)

    # Borrow a book using its title
    def borrow_book(self, title):
        # Search through the library's books
        for book in self.__books:

            # Compare titles without case sensitivity
            if book.title.lower() == title.lower():

                # Check if the book is available
                if not book.is_borrowed:
                    # Mark the book as borrowed
                    book.is_borrowed = True

                    # Confirm borrowing
                    print(f"You have successfully borrowed '{book.title}'.")
                    return

                else:
                    # Inform the user that it is already borrowed
                    print(f"Sorry, '{book.title}' is currently borrowed.")
                    return

        # This executes if the title was not found
        print(f"Book '{title}' not found in library.")

# Create the library object
city_library = Library("City Central Library")

# Create Book objects
b1 = Book("Python Crash Course", "Eric Matthes")
b2 = Book("Clean Code", "Robert C. Martin")

# Add the books to the library
city_library.add_book(b1)
city_library.add_book(b2)

# Display the catalog
city_library.show_books()

# Borrow a book
city_library.borrow_book("Clean Code")

# Try borrowing the same book again
city_library.borrow_book("Clean Code")

# Display the catalog after borrowing
city_library.show_books()
````

### Expected Output
```text
Added: Python Crash Course
Added: Clean Code

--- City Central Library Catalog ---
Python Crash Course by Eric Matthes [Available]
Clean Code by Robert C. Martin [Available]
You have successfully borrowed 'Clean Code'.
Sorry, 'Clean Code' is currently borrowed.

--- City Central Library Catalog ---
Python Crash Course by Eric Matthes [Available]
Clean Code by Robert C. Martin [Borrowed]
```

### Project Walkthrough

1. `Book` represents an individual book.
2. `Library` represents the library and manages its collection.
3. The library's `__books` list is name-mangled to indicate internal use.
4. `add_book()` adds a Book object.
5. `show_books()` prints each object using `__str__()`.
6. `borrow_book()` searches for a book, checks availability, and updates its status.
7. The same book cannot be borrowed twice while marked as borrowed.

# 9. Classroom Exercise: SmartPhone

## Student Prompt

Create a class called `SmartPhone` with the following requirements.

### Attributes
- `brand`
- `model`
- Private `__battery` attribute set to 100 by default.

### Methods
1. `use_app(minutes)` — reduces battery by `minutes * 0.5`; battery should never go below 0.
2. `charge()` — resets battery to 100.
3. `status()` — displays brand, model, and current battery percentage.

### Bonus Challenge
Create a child class `GamingPhone` that inherits from `SmartPhone` and:
- Adds a `cooling_mode` attribute.
- Implements a `boost_performance()` method.

Try implementing the solution before proceeding.


## Exercise Solution: SmartPhone and GamingPhone


````python
# Define the SmartPhone class
class SmartPhone:

    # Constructor with battery initialized to 100
    def __init__(self, brand, model):
        # Store the phone brand
        self.brand = brand

        # Store the phone model
        self.model = model

        # Private battery attribute
        self.__battery = 100

    # Method to simulate using an app
    def use_app(self, minutes):
        # Calculate battery usage
        battery_used = minutes * 0.5

        # Reduce battery but prevent it from becoming negative
        self.__battery = max(0, self.__battery - battery_used)

        # Display battery usage information
        print(f"Used app for {minutes} minutes. Battery: {self.__battery}%")

    # Method to charge the phone
    def charge(self):
        # Reset battery to full
        self.__battery = 100

        # Display charging confirmation
        print("Phone charged to 100%.")

    # Method to display phone status
    def status(self):
        # Display phone information and battery level
        print(f"Brand: {self.brand}")
        print(f"Model: {self.model}")
        print(f"Battery: {self.__battery}%")

# Define the GamingPhone child class
class GamingPhone(SmartPhone):

    # Constructor
    def __init__(self, brand, model, cooling_mode):
        # Initialize inherited attributes using the parent constructor
        super().__init__(brand, model)

        # Store child-specific cooling mode
        self.cooling_mode = cooling_mode

    # Child-specific performance method
    def boost_performance(self):
        # Display performance boost information
        print(f"Performance boost activated! Cooling mode: {self.cooling_mode}")

# Create a normal smartphone
phone = SmartPhone("Samsung", "Galaxy A55")

# Display initial status
phone.status()

# Simulate app usage
phone.use_app(60)

# Charge the phone
phone.charge()

# Display updated status
phone.status()

# Create a gaming phone
gaming_phone = GamingPhone("ASUS", "ROG Phone", "Liquid Cooling")

# Display inherited phone status
gaming_phone.status()

# Use the child-specific method
gaming_phone.boost_performance()
````

### Expected Output
```text
Brand: Samsung
Model: Galaxy A55
Battery: 100%
Used app for 60 minutes. Battery: 70.0%
Phone charged to 100%.
Brand: Samsung
Model: Galaxy A55
Battery: 100%
Brand: ASUS
Model: ROG Phone
Battery: 100%
Performance boost activated! Cooling mode: Liquid Cooling
```

### Solution Explanation

- `SmartPhone` is the parent class.
- `__battery` is an encapsulated attribute.
- `use_app()` calculates battery consumption and uses `max()` to prevent a negative battery value.
- `charge()` resets the battery to 100.
- `status()` displays the phone information.
- `GamingPhone(SmartPhone)` inherits from SmartPhone.
- `super().__init__()` initializes the parent attributes.
- `cooling_mode` is an additional child attribute.
- `boost_performance()` is a child-specific method.
