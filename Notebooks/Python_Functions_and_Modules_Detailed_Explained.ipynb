# 1. What is a function?

A **function** is a named, reusable block of code that performs a particular task. The code inside a function runs when the function is called.

### Why do we use functions?
- **Reusability:** Write logic once and call it many times.
- **Modularity:** Divide a large program into smaller tasks.
- **Readability:** Give meaningful names to operations.
- **Maintenance:** Update logic in one place instead of changing repeated code.

### General syntax

```python
def function_name(parameters):
    # Statements that perform the task
    return result  # optional
```

- `def` starts a function definition.
- `function_name` is the name used to call the function.
- `parameters` are input variables; they are optional.
- The indented block is the function body.
- `return` sends a result back to the caller; it is optional.

## 1.1 A function without parameters

A function does not always need input. A function without parameters performs a task using values already available inside its body.

In this example, `hello()` prints a greeting. The function has no parameters, and it does not return a value.

````python
# Define a function that does not accept any parameters.
def hello():
    # Display a greeting message.
    print("Hello, welcome to Python functions!")

# Call the function so that its body executes.
hello()
````

### Code explanation
1. `def hello():` defines a function named `hello`.
2. The indented `print()` statement belongs to the function body.
3. `hello()` calls the function. Defining a function alone does not execute it.

### Expected output
```text
Hello, welcome to Python functions!
```

## 1.2 A function with parameters

A **parameter** is a variable written in the function definition. An **argument** is the actual value supplied when the function is called.

Here, `name` is a parameter. The string `"Prithivi"` is the argument.

````python
# Define a function with one parameter named name.
def greet(name):
    # Use the received value to create a personalised greeting.
    print("Hello, {}".format(name))

# Call the function and pass an argument.
greet("Prithivi")

# Call the same function with a different argument.
greet("Aarav")
````

### Code explanation
- `def greet(name):` creates a function that expects one value.
- `format(name)` inserts the received value into the message.
- Each function call supplies a different argument, so the same function can be reused.

### Expected output
```text
Hello, Prithivi
Hello, Aarav
```

# 2. Types of function arguments

Python supports several ways of passing arguments. The presentation introduces required arguments, keyword arguments, default arguments and variable-length arguments.

## 2.1 Required (positional) arguments

Required arguments must be supplied when the function is called. With positional arguments, Python matches values to parameters according to their position.

If a required argument is missing, Python raises a `TypeError`. If values are supplied in the wrong order, they may be assigned to the wrong parameters.

````python
# Define a function that requires an employee ID and employee name.
def display_employee(employee_id, employee_name):
    # Display both received values.
    print("Employee ID:", employee_id)
    print("Employee Name:", employee_name)

# Supply both required arguments in the correct order.
display_employee("Ram", 10)
````

### Code explanation
- The function definition contains two required parameters.
- The first argument (`10`) is assigned to `employee_id`.
- The second argument (`"Ram"`) is assigned to `employee_name`.
- Both values must be supplied for this call.

### Expected output
```text
Employee ID: 10
Employee Name: Ram
```

**Try it:** Remove one argument from the function call and run the cell. Observe the `TypeError`.

## 2.2 Keyword arguments

A **keyword argument** is passed by writing the parameter name followed by `=` and its value. Python matches the value using the parameter name, so keyword arguments can be supplied in a different order from the function definition.

Keyword arguments improve readability when a function has several parameters.

````python
# Define a function with two parameters.
def display_employee(employee_id, employee_name):
    print("Employee ID:", employee_id)
    print("Employee Name:", employee_name)

# Pass values by naming their parameters.
# The order is different from the function definition, but names match.
display_employee(employee_name="Dhruv", employee_id=20)
````

### Code explanation
- `employee_name="Dhruv"` assigns `"Dhruv"` to the parameter named `employee_name`.
- `employee_id=20` assigns `20` to `employee_id`.
- Because the arguments are named, their order does not determine the assignment.

### Expected output
```text
Employee ID: 20
Employee Name: Dhruv
```

## 2.3 Default arguments

A **default argument** has a value assigned in the function definition. If the caller does not provide a value for that parameter, Python uses the default.

Default parameters are useful when a value is commonly the same, but callers should still be able to override it.

**Important rule:** Parameters with default values should come after parameters without default values.

````python
# 'designation' and 'age' have default values.
def display_person(name, designation="Officer", age=30):
    # Display the supplied name and the effective optional values.
    print(f"{name} is an {designation} and age is {age}")

# Only the required parameter is supplied; defaults are used.
display_person("Pranav")

# The designation is overridden; the default age is retained.
display_person("Pranav", "Pilot")

# Use a keyword argument to override age while keeping the default designation.
display_person("Pranav", 30)
````

### Code explanation
1. `name` is required because it has no default value.
2. `designation="Officer"` is used when no designation is passed.
3. `age=30` is used when no age is passed.
4. The second call replaces the default designation with `"Pilot"`.
5. The third call uses a keyword argument to set `age` while skipping `designation`.

### Expected output
```text
Pranav is an Officer and age is 30
Pranav is an Pilot and age is 30
Pranav is an Officer and age is 32
```

**Note:** For grammatically polished output, you could change the message to `"{name} works as a {designation}..."`.

## 2.4 Mixing positional and keyword arguments

Python allows positional and keyword arguments in the same function call. Positional arguments must come before keyword arguments.

This is valid:
```python
display_person("Pranav", age=32)
```

This is invalid:
```python
# display_person(name="Pranav", 32)
```
The positional argument appears after a keyword argument.

````python
# Reuse a function with one required and two default parameters.
def display_person(name, designation="Officer", age=30):
    print(f"Name: {name} | Designation: {designation} | Age: {age}")

# First argument is positional; age is passed by keyword.
display_person("Pranav", age=32)
````

### Expected output
```text
Name: Pranav | Designation: Officer | Age: 32
```

# 3. Variable-length arguments

Sometimes the number of arguments is not known when a function is written. Python supports variable-length arguments for this situation.

- `*args` collects extra positional arguments into a **tuple**.
- `**kwargs` collects extra keyword arguments into a **dictionary**.

The names `args` and `kwargs` are conventions. The `*` and `**` symbols provide the special behaviour.

## 3.1 `*args` — variable number of positional arguments

Use `*args` when a function should accept zero or more positional arguments. Inside the function, `args` behaves like a tuple.

This example calculates the total of any number of values.

````python
# The * before numbers collects all extra positional arguments into a tuple.
def add_numbers(*numbers):
    # Start the running total at zero.
    total = 0

    # Visit each supplied number one by one.
    for number in numbers:
        # Add the current number to the running total.
        total += number

    # Return the final total to the caller.
    return total

# The same function supports different numbers of arguments.
print(add_numbers())
print(add_numbers(5))
print(add_numbers(5, 10))
print(add_numbers(5, 10, 15, 20))
````

### Code explanation
- `*numbers` collects all positional arguments into a tuple.
- The `for` loop processes every value in that tuple.
- `total += number` adds the current value to the running total.
- `return total` sends the calculated result back.
- With no arguments, the loop does not run and the result remains `0`.

### Expected output
```text
0
5
15
50
```

## 3.2 `**kwargs` — variable number of keyword arguments

Use `**kwargs` when a function should accept an unknown number of named arguments. Inside the function, `kwargs` is a dictionary containing the parameter names and their values.

This is useful when records may contain different fields.

````python
# The ** before details collects keyword arguments into a dictionary.
def display_employee_details(**details):
    # items() gives each key and its corresponding value.
    for key, value in details.items():
        # Print the field name and its value.
        print(f"{key}: {value}")

# Pass three named values to the function.
display_employee_details(employee_id=1001, name="Hillary", age=30)

print("---")

# The same function can receive a different set of fields.
display_employee_details(name="Asha", department="IT")
````

### Code explanation
- `**details` collects named arguments into a dictionary.
- `details.items()` returns key-value pairs.
- The loop prints every field without needing to know the field names in advance.

### Expected output
```text
employee_id: 1001
name: Hillary
age: 30
---
name: Asha
department: IT
```

## 3.3 Using regular parameters with `*args` and `**kwargs`

A function can combine regular parameters, `*args` and `**kwargs`.

A common parameter order is:
```python
def function_name(required, *args, **kwargs):
    ...
```
The regular parameter receives its expected value, `*args` collects additional positional values, and `**kwargs` collects additional named values.

````python
# 'course' is a regular parameter.
# '*participants' collects extra positional values.
# '**details' collects extra keyword values.
def show_training_batch(course, *participants, **details):
    print("Course:", course)

    # Display all participant names supplied positionally.
    print("Participants:", participants)

    # Display optional batch details supplied as keywords.
    print("Batch details:", details)

# Call the function with all three argument styles.
show_training_batch(
    "Python",
    "Aman",
    "Neha",
    trainer="Mausomi",
    duration_days=5
)
````

### Expected output
```text
Course: Python
Participants: ('Aman', 'Neha')
Batch details: {'trainer': 'Mausomi', 'duration_days': 5}
```

# 4. Returning values from a function

`return` sends a result from a function back to the statement that called it. This is different from `print()`:
- `print()` displays information on the screen.
- `return` makes a value available to the rest of the program.

A function can return one value, multiple values, or no explicit value. If there is no `return` statement, Python returns `None`.

## 4.1 Returning a single value

````python
# Define a function that calculates the square of a number.
def calculate_square(number):
    # Calculate the square and return it to the caller.
    return number ** 2

# Store the returned result in a variable.
result = calculate_square(5)

# Display the stored result.
print("Square:", result)
````

### Code explanation
- `number ** 2` raises the input number to the power of two.
- `return` sends the result to the caller.
- The function call's returned value is assigned to `result`.

### Expected output
```text
Square: 25
```

## 4.2 Returning multiple values

A Python function can return multiple values separated by commas. Python packages these values into a tuple.

The caller can receive the tuple in two common ways:
1. **Unpacking:** Assign each returned value to a separate variable.
2. **Indexing:** Store the tuple and access values using indexes.

````python
# Calculate and return both the square and cube of a number.
def square_and_cube(number):
    # Calculate the square.
    square = number ** 2

    # Calculate the cube.
    cube = number ** 3

    # Return both values; Python returns them together as a tuple.
    return square, cube

# Method 1: Unpack the returned tuple into two variables.
square_value, cube_value = square_and_cube(5)

print("Square of 5:", square_value)
print("Cube of 5:", cube_value)

# Method 2: Store the returned tuple and access its elements by index.
answer = square_and_cube(4)
print("Returned tuple:", answer)
print("Square of 4:", answer[0])
print("Cube of 4:", answer[1])
````

### Code explanation
- `return square, cube` returns a tuple such as `(25, 125)`.
- `square_value, cube_value = ...` unpacks the two tuple elements.
- `answer[0]` accesses the first element; `answer[1]` accesses the second.
- Python tuple indexes start at `0`.

### Expected output
```text
Square of 5: 25
Cube of 5: 125
Returned tuple: (16, 64)
Square of 4: 16
Cube of 4: 64
```

# 5. How arguments behave when objects are passed to functions

When a variable is passed to a function, Python passes an object reference. The effect depends on what the function does with that object.

- **Mutable objects**, such as lists and dictionaries, can be changed in place. Such a change can be visible outside the function.
- **Immutable objects**, such as integers, strings and tuples, cannot be changed in place. Reassigning a parameter to a new value does not reassign the caller's variable.

Some introductory materials describe these examples as call by reference and call by value. A more precise description for Python is **object-reference (or call-by-sharing) semantics**.

## 5.1 Modifying a mutable object inside a function

A list is mutable. If a function changes one of its elements, it changes the same list object that the caller refers to.

````python
# Create a list outside the function.
fruits = ["apple", "orange", "kiwi"]
# Define a function that receives a reference to the list.
def change_fruits(fruit_list):
    # Modify the element at index 1 in the existing list.
    fruit_list[1] = "strawberry"
# Display the list before the function call.
print("Before function call:", fruits)
# Pass the list to the function.
change_fruits(fruits)
# Display the list after the function call.
print("After function call:", fruits)
````

### Code explanation
- `fruits` refers to a list object.
- `fruit_list` refers to that same list while the function runs.
- Assigning a new value to index `1` modifies the existing list in place.
- Therefore, the change is visible through `fruits` after the function call.

### Expected output
```text
Before function call: ['apple', 'orange', 'kiwi']
After function call: ['apple', 'strawberry', 'kiwi']
```

## 5.2 Reassigning a function parameter

Reassigning a parameter to a new list does not replace the caller's variable. The parameter becomes bound to a different list object inside the function.

````python
# Create a list in the main program.
values = [1, 2]

# Define a function that reassigns its local parameter.
def replace_list(items):
    # This assignment makes 'items' refer to a new list.
    # It does not replace the caller's variable 'values'.
    items = [10, 20]

    # Display the new list referenced by the local parameter.
    print("Inside function:", items)

# Display the original list before the call.
print("Before function call:", values)

# Pass the original list to the function.
replace_list(values)

# The caller's variable still refers to the original list.
print("After function call:", values)
````

### Code explanation
- At the start of the function, `items` refers to the same list as `values`.
- `items = [10, 20]` rebinds only the local name `items` to a new list.
- The original list is not modified.
- Therefore, `values` remains `[1, 2]`.

### Expected output
```text
Before function call: [1, 2]
Inside function: [10, 20]
After function call: [1, 2]
```

# 6. Practical programs using user-defined functions

The following examples apply functions to common business and classroom scenarios.

## 6.1 Program: Calculate simple interest

Simple interest is calculated using:

\[
SI = \frac{P \times R \times T}{100}
\]

where:
- `P` is the principal amount.
- `R` is the annual interest rate expressed as a percentage.
- `T` is the time in years.

The function below accepts the three inputs and returns the interest amount.

````python
# Define a reusable function to calculate simple interest.
def calculate_simple_interest(principal, annual_rate, time_years):
    # Apply the simple-interest formula.
    interest = (principal * annual_rate * time_years) / 100

    # Return the calculated interest.
    return interest

# Example 1: Calculate interest on a loan.
loan_amount = 100000
loan_rate = 5
loan_duration = 3

# Call the function and store the returned interest.
loan_interest = calculate_simple_interest(
    loan_amount, loan_rate, loan_duration
)

# Display the interest with commas and two decimal places.
print(f"Loan interest payable: Rs. {loan_interest:,.2f}")

# Example 2: Calculate interest on a deposit.
deposit_amount = 50000
deposit_rate = 3
deposit_duration = 1

deposit_interest = calculate_simple_interest(
    deposit_amount, deposit_rate, deposit_duration
)

print(f"Deposit interest earned: Rs. {deposit_interest:,.2f}")
````

### Code explanation
- The function receives principal, annual rate and time in years.
- The formula multiplies the three inputs and divides by `100`.
- The returned value can be reused for loans, deposits or other examples.
- `:,.2f` formats a number with a thousands separator and two digits after the decimal point.

### Expected output
```text
Loan interest payable: Rs. 15,000.00
Deposit interest earned: Rs. 1,500.00
```

## 6.2 Program: Validate an account or patient ID

Validation functions check whether input follows a rule before it is accepted.

This example checks whether an ID contains between 5 and 16 characters. It is a simple length check, not a complete identity or security validation.

````python
# Define a function that checks the length of an ID string.
def is_valid_patient_id(patient_id):
    # Set the allowed minimum and maximum lengths.
    min_length = 5
    max_length = 16

    # Check whether the ID length is within the allowed range.
    if min_length <= len(patient_id) <= max_length:
        return True
    else:
        return False

# Create sample IDs for testing.
patient1 = "P-101-ABC"
patient2 = "AB"
patient3 = "PATIENT-2026-XYZ"

# Call the validation function for each sample.
print(f"{patient1} is valid: {is_valid_patient_id(patient1)}")
print(f"{patient2} is valid: {is_valid_patient_id(patient2)}")
print(f"{patient3} is valid: {is_valid_patient_id(patient3)}")
````

### Code explanation
- `len(patient_id)` returns the number of characters in the string.
- The chained comparison checks both the lower and upper limits.
- The function returns `True` when the length is allowed; otherwise, it returns `False`.

### Expected output
```text
P-101-ABC is valid: True
AB is valid: False
PATIENT-2026-XYZ is valid: True
```

# 7. Lambda functions

A **lambda function** is a small anonymous function created using the `lambda` keyword. It is often used for short operations where defining a full function with `def` would be unnecessary.

### Syntax
```python
lambda arguments: expression
```

- A lambda can accept any number of arguments.
- It contains one expression.
- The expression's result is returned automatically.
- A lambda is not a replacement for every normal function; use `def` for multi-step or reusable logic.

## 7.1 Basic lambda example

````python
# Create a lambda function that doubles its input.
double = lambda number: number * 2

# Call the lambda function just like a normal function.
print(double(5))
print(double(12))
````

### Code explanation
- `lambda number: number * 2` accepts one argument named `number`.
- The expression multiplies the argument by two.
- The lambda is assigned to `double`, so it can be called using that name.

### Expected output
```text
10
24
```

## 7.2 Lambda with multiple arguments

````python
# Define a lambda expression that multiplies two values.
multiply = lambda first, second: first * second

# Pass two arguments to the lambda.
print(multiply(6, 7))
````

### Expected output
```text
42
```

## 7.3 Lambda with a conditional expression

A conditional expression can be used inside a lambda to choose between two results. This example classifies a number as even or odd.

````python
# Return an appropriate label based on whether the number is divisible by 2.
odd_or_even = lambda number: "Even" if number % 2 == 0 else "Odd"

# Test the lambda with different values.
print(odd_or_even(8))
print(odd_or_even(11))
````

### Expected output
```text
Even
Odd
```

# 8. `map()` function

The built-in `map()` function applies a function to every item in an iterable, such as a list or tuple.

### Syntax
```python
map(function, iterable)
```

- `function` is the operation to apply.
- `iterable` supplies the input values.
- `map()` returns a map iterator, not a list.
- Use `list()` or another collection constructor when you want to display or store all results immediately.

Use `map()` when every input item needs the same transformation.

## 8.1 Use `map()` to classify numbers

````python
# Define a lambda that returns an even/odd message for one number.
classify_number = lambda number: (
    f"{number} is even" if number % 2 == 0 else f"{number} is odd"
)

# Create the input list.
numbers = [2, 4, 5, 6, 7, 8, 10, 11]

# Apply the lambda to every item using map().
mapped_result = map(classify_number, numbers)

# Convert the map iterator to a list so that all results are displayed.
print(list(mapped_result))
````

### Code explanation
1. The lambda receives one number and returns a string.
2. `map(classify_number, numbers)` applies it to each list item.
3. `map()` produces an iterator.
4. `list()` consumes the iterator and creates a list of results.

### Expected output
```text
['2 is even', '4 is even', '5 is odd', '6 is even',
 '7 is odd', '8 is even', '10 is even', '11 is odd']
```

## 8.2 Use `map()` to multiply corresponding elements of two lists

When multiple iterables are supplied, `map()` passes corresponding items to the function. It stops when the shortest iterable is exhausted.

````python
# Create two lists containing corresponding values.
prices = [10, 20, 30]
quantities = [2, 3, 5]

# Multiply each price by its matching quantity.
line_totals = map(lambda price, quantity: price * quantity,
                  prices, quantities)

# Convert the iterator into a list and display the result.
print(list(line_totals))
````

### Expected output
```text
[20, 60, 150]
```

### Explanation
- First call: `10 * 2 = 20`
- Second call: `20 * 3 = 60`
- Third call: `30 * 5 = 150`

## 8.3 Practical program: Convert pounds to kilograms

One pound is approximately `0.453592` kilograms. `map()` can apply this conversion to every weight in a list.

````python
# Store sample weights in pounds.
weights_lb = [150, 205, 120, 185]

# Convert each weight to kilograms using map() and a lambda.
weights_kg = list(
    map(lambda weight: weight * 0.453592, weights_lb)
)

# Display the original weights.
print("Weights in pounds:", weights_lb)

# Format converted values to two decimal places for readability.
formatted_weights = [f"{weight:.2f}" for weight in weights_kg]
print("Weights in kilograms:", formatted_weights)
````

### Expected output
```text
Weights in pounds: [150, 205, 120, 185]
Weights in kilograms: ['68.04', '92.99', '54.43', '83.91']
```

# 9. `filter()` function

The built-in `filter()` function selects items from an iterable based on a condition.

### Syntax
```python
filter(function, iterable)
```

The function should return a truthy or falsy result:
- If it returns `True`, the item is included.
- If it returns `False`, the item is excluded.

Like `map()`, `filter()` returns an iterator. Convert it to a list to display all selected values.

## 9.1 Filter numbers greater than 50

````python
# Create a list of numbers.
numbers = [45, 23, 67, 98, 34, 23, 38]

# Keep only the values for which the condition number > 50 is True.
filtered_numbers = filter(lambda number: number > 50, numbers)

# Convert the filter iterator into a list.
print(list(filtered_numbers))
````

### Code explanation
- The lambda tests each number.
- `67` and `98` satisfy the condition.
- Values that do not satisfy the condition are discarded.
- `list()` displays the selected values.

### Expected output
```text
[67, 98]
```

## 9.2 Practical program: Filter high blood-pressure readings

This example uses a simple threshold to demonstrate filtering. It is only a programming exercise, not a clinical risk assessment or medical recommendation.

````python
# Store sample systolic blood-pressure readings.
bp_readings = [125, 155, 130, 142, 118, 160]

# Select readings greater than the example threshold of 140.
high_readings = list(
    filter(lambda reading: reading > 140, bp_readings)
)

# Display all readings and the selected readings.
print("All readings:", bp_readings)
print("Readings above 140:", high_readings)
````

### Expected output
```text
All readings: [125, 155, 130, 142, 118, 160]
Readings above 140: [155, 142, 160]
```

# 10. `reduce()` function

The `reduce()` function repeatedly combines items from an iterable and produces one accumulated result.

Unlike `map()` and `filter()`, `reduce()` is not a built-in function in the global namespace. It must be imported from the `functools` module.

### Syntax
```python
from functools import reduce
reduce(function, iterable)
```

The combining function usually accepts two values:
- An accumulated result.
- The next item from the iterable.

The returned result is passed into the next application of the function.

## 10.1 Use `reduce()` to calculate a total

````python
# Import reduce from Python's functools module.
from functools import reduce

# Store individual costs.
prescription_costs = [35.50, 12.75, 89.99, 5.00, 25.40]

# Add values cumulatively to calculate the grand total.
total_cost = reduce(lambda running_total, next_cost:
                    running_total + next_cost,
                    prescription_costs)

# Display the original costs and the final total.
print("Individual costs:", prescription_costs)
print(f"Total cost: Rs. {total_cost:.2f}")
````

### Code explanation
- `reduce()` starts with the first two values and applies the lambda.
- It then combines the accumulated result with the next value.
- This continues until all values have been processed.
- The final accumulated value is the sum of the list.

### Expected output
```text
Individual costs: [35.5, 12.75, 89.99, 5.0, 25.4]
Total cost: Rs. 168.64
```

## 10.2 Trace how `reduce()` works

````python
# Import reduce if this cell is run independently.
from functools import reduce

# Define a function that prints each combination step.
def show_addition_step(first, second):
    # Calculate the intermediate result.
    result = first + second

    # Display the two values being combined and their result.
    print(f"{first} + {second} = {result}")

    # Return the result so reduce() can use it in the next step.
    return result

# Reduce the list to one sum while displaying each step.
final_sum = reduce(show_addition_step, [10, 20, 30, 40])

print("Final sum:", final_sum)
````

## 10.3 Reduce with integer division

The presentation also demonstrates cumulative integer division. The operation is applied from left to right, so each intermediate result becomes the first argument in the next step.

````python
# Import reduce from functools.
from functools import reduce

# Apply floor division cumulatively to the sequence.
numbers = [1000, 2, 4, 5]
answer = reduce(lambda first, second: first // second, numbers)

# Display the final reduced value.
print(answer)
````

# 11. Comparing `map()`, `filter()` and `reduce()`

| Function | Main purpose | Result |
|---|---|---|
| `map()` | Transform every item | Iterator of transformed items |
| `filter()` | Keep items that pass a condition | Iterator of selected items |
| `reduce()` | Combine items cumulatively | One final value |

A quick way to remember:
- **map:** change each item.
- **filter:** select some items.
- **reduce:** combine items into one result.

````python
# Import reduce for the third operation.
from functools import reduce

# A sample list used by all three operations.
numbers = [1, 2, 3, 4, 5]

# MAP: multiply every number by 10.
mapped = list(map(lambda number: number * 10, numbers))

# FILTER: keep only even numbers.
filtered = list(filter(lambda number: number % 2 == 0, numbers))

# REDUCE: add all numbers to create one total.
total = reduce(lambda first, second: first + second, numbers)

# Display each result.
print("map result:", mapped)
print("filter result:", filtered)
print("reduce result:", total)
````

### Expected output
```text
map result: [10, 20, 30, 40, 50]
filter result: [2, 4]
reduce result: 15
```

# 13. Creating your own module

You can create a custom module by saving Python code in a `.py` file. Functions and variables defined in that file can then be imported into another Python program.

For a practical demonstration, we will create:
1. `my_utils.py` — contains a temperature conversion function.
2. A separate code cell — imports the module and uses the function.

In Jupyter, the `%%writefile` command is a notebook magic that writes the cell contents to a file.

## 13.1 Create `my_utils.py`

````python
%%writefile my_utils.py
# This file is a custom Python module named my_utils.py.

def celsius_to_fahrenheit(temp_c):
    # Convert Celsius to Fahrenheit using the standard formula.
    temp_f = (temp_c * 9 / 5) + 32

    # Return the converted temperature to the caller.
    return temp_f

def fahrenheit_to_celsius(temp_f):
    # Convert Fahrenheit to Celsius.
    temp_c = (temp_f - 32) * 5 / 9

    # Return the converted temperature.
    return temp_c
````

### Explanation
- `%%writefile my_utils.py` saves the cell as a Python file.
- The file contains two reusable conversion functions.
- The module does not need to contain a main program; it can simply provide reusable tools.

### Expected output
Jupyter normally displays a confirmation similar to:
```text
Writing my_utils.py
```

## 13.2 Import and use the custom module

````python
# Import the custom module created in the previous cell.
import my_utils

# Set a temperature in Celsius.
temperature_c = 25

# Call the conversion function through the module name.
temperature_f = my_utils.celsius_to_fahrenheit(temperature_c)

# Display both temperatures.
print(f"{temperature_c}°C = {temperature_f:.1f}°F")

# Use the second function from the same module.
print(f"98.6°F = {my_utils.fahrenheit_to_celsius(98.6):.1f}°C")
````

### Code explanation
- `import my_utils` loads the custom module.
- `my_utils.celsius_to_fahrenheit(...)` calls the function defined in that file.
- The dot (`.`) accesses a name inside the module.
- The same module can provide multiple reusable functions.

### Expected output

# 14. The `if __name__ == "__main__"` pattern

A Python file can be run directly or imported by another file. The special condition below helps separate code intended to run directly from code that should only be available when imported.

```python
if __name__ == "__main__":
    # Statements here run when this file is executed directly.
```

- When a file is run directly, Python sets `__name__` to `"__main__"`.
- When a file is imported as a module, `__name__` is set to the module's name.
- This pattern prevents test or demonstration code from running automatically during import.

````python
%%writefile calculator_tools.py
# A reusable function that can be imported by other programs.
def add(first, second):
    return first + second

# This block runs only when calculator_tools.py is executed directly.
if __name__ == "__main__":
    # Demonstrate the add() function when running this file itself.
    print("Demo result:", add(10, 20))
````

### Explanation
- `add()` is defined at module level, so another program can import it.
- The `if` block is a direct-execution guard.
- When the file is imported, the function is available but the demo print statement does not run.

### Expected output
When the file is executed directly from a terminal:
```text
Demo result: 30
```
When imported from another Python file, the demo output is not printed automatically.

````python
# Import the function from the custom module.
from calculator_tools import add

# Call the imported function.
result = add(15, 25)

# Display the result from the current notebook.
print("Imported function result:", result)
````

### Expected output
```text
Imported function result: 40
```

# 15. Organising a small project with multiple modules

A larger program can be split into focused modules. For example:

```text
banking_project/
    main.py
    loan_calculations.py
    customer_validators.py
```

- `loan_calculations.py` can contain loan-related functions.
- `customer_validators.py` can contain customer-data validation functions.
- `main.py` can import and coordinate those functions.

The next cells demonstrate this structure using files in the current notebook directory.

## 15.1 Create a loan calculation module

````python
%%writefile loan_calculations.py
# Reusable loan-related functions.

def simple_interest(principal, rate, years):
    # Calculate simple interest and return the result.
    return (principal * rate * years) / 100

def total_amount(principal, rate, years):
    # Calculate interest by reusing the simple_interest() function.
    interest = simple_interest(principal, rate, years)

    # Return principal plus interest.
    return principal + interest
````

## 15.2 Create a customer validation module

````python
%%writefile customer_validators.py
# Reusable customer-data validation functions.

def is_valid_customer_name(name):
    # A simple check: the name must contain non-whitespace characters.
    return isinstance(name, str) and len(name.strip()) > 0

def is_valid_account_number(account_number):
    # Convert the account number to text for a simple length check.
    account_number = str(account_number)

    # Accept account numbers containing 8 to 12 characters.
    return account_number.isdigit() and 8 <= len(account_number) <= 12
````

## 15.3 Import and use both modules

````python
# Import the two custom modules.
import loan_calculations
import customer_validators

# Calculate a sample loan amount.
principal = 200000
rate = 8
years = 2

interest = loan_calculations.simple_interest(principal, rate, years)
total = loan_calculations.total_amount(principal, rate, years)

# Validate sample customer information.
customer_name = "Ananya"
account_number = "1234567890"

# Display calculation and validation results.
print(f"Interest: Rs. {interest:,.2f}")
print(f"Total amount: Rs. {total:,.2f}")
print("Customer name valid:", customer_validators.is_valid_customer_name(customer_name))
print("Account number valid:", customer_validators.is_valid_account_number(account_number))
````

### Expected output
```text
Interest: Rs. 32,000.00
Total amount: Rs. 232,000.00
Customer name valid: True
Account number valid: True
```

**Project takeaway:** Separating calculations and validations into different modules makes the project easier to understand, test and maintain.

# 16. Common errors and debugging tips

| Error or issue | Why it happens | How to fix it |
|---|---|---|
| `TypeError: missing required positional argument` | A required argument was not supplied | Pass all required arguments |
| `TypeError: got multiple values for argument` | A parameter received a value positionally and by keyword | Supply the value only once |
| `NameError` | A function or variable name is misspelled or not defined | Check spelling and definition order |
| `ModuleNotFoundError` | Python cannot locate the module | Check the filename and working directory |
| `TypeError` with `map()` or `filter()` | The function and iterable arguments are not compatible | Check the function's expected inputs |
| `reduce` is not defined | It was not imported | Use `from functools import reduce` |

### Debugging habits
- Read the final line of the traceback first; it usually identifies the exception.
- Check the function definition and the exact function call.
- Print values temporarily to inspect what is being passed.
- Test functions with normal, boundary and unexpected inputs.
- Use meaningful parameter and variable names.

# 18. Mini project — Student result processor

## Problem statement
Create a small result-processing program using functions. The program should:
1. Accept a student's name and marks.
2. Calculate the total and average.
3. Assign a result category based on the average.
4. Return multiple values from a function.
5. Use `map()` to create a list of averages for several students.

For this classroom example, the categories are:
- `Distinction`: average is 75 or above.
- `First Class`: average is 60 or above.
- `Pass`: average is 40 or above.
- `Needs Improvement`: average is below 40.

These thresholds are example rules for the exercise.

````python
# Define a function that calculates a student's result.
def process_student(name, marks):
    # Calculate the sum of all subject marks.
    total = sum(marks)

    # Calculate the average; marks is expected to contain at least one value.
    average = total / len(marks)

    # Select a category based on the average.
    if average >= 75:
        category = "Distinction"
    elif average >= 60:
        category = "First Class"
    elif average >= 40:
        category = "Pass"
    else:
        category = "Needs Improvement"

    # Return multiple values as a tuple.
    return name, total, average, category

# Create sample student records.
students = [
    ("Aman", [80, 75, 90]),
    ("Neha", [65, 70, 60]),
    ("Ravi", [35, 45, 30])
]

# Process and display each student record.
for student_name, student_marks in students:
    name, total, average, category = process_student(
        student_name, student_marks
    )

    print(f"Student: {name}")
    print(f"Total: {total}")
    print(f"Average: {average:.2f}")
    print(f"Category: {category}")
    print("-" * 30)
````
