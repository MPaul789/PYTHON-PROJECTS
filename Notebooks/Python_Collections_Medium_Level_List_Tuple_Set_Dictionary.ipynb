# 1. Lists

## Introduction
A **list** is an ordered, mutable collection that can store multiple values. Lists allow duplicate values and can contain elements of different data types.

### Key Characteristics
- Ordered: items maintain their position.
- Mutable: items can be changed after creation.
- Allows duplicates.
- Supports indexing and slicing.
- Can store mixed data types and nested collections.

### Common Operations
| Operation | Description |
|---|---|
| `append()` | Adds one item at the end |
| `extend()` | Adds multiple items |
| `insert()` | Inserts an item at a specific position |
| `remove()` | Removes the first matching value |
| `pop()` | Removes and returns an item |
| `sort()` | Sorts the list in place |
| `reverse()` | Reverses the list in place |
| `index()` | Finds the position of a value |
| `count()` | Counts occurrences |


````python
# Create a list of student marks
marks = [78, 85, 92, 67, 85, 90]

# Display the original list
print("Original marks:", marks)

# Add a new mark at the end
marks.append(88)
print("After append:", marks)

# Add multiple marks
marks.extend([76, 95])
print("After extend:", marks)

# Insert a mark at index 2
marks.insert(2, 81)
print("After insert:", marks)

# Remove the first occurrence of 85
marks.remove(85)
print("After remove:", marks)

# Remove and return the last item
removed_mark = marks.pop()
print("Removed mark:", removed_mark)
print("After pop:", marks)
````

### Expected Output
```text
Original marks: [78, 85, 92, 67, 85, 90]
After append: [78, 85, 92, 67, 85, 90, 88]
After extend: [78, 85, 92, 67, 85, 90, 88, 76, 95]
After insert: [78, 85, 81, 92, 67, 85, 90, 88, 76, 95]
After remove: [78, 85, 81, 92, 67, 90, 88, 76, 95]
Removed mark: 95
After pop: [78, 85, 81, 92, 67, 90, 88, 76]
```


## 1.1 List Indexing and Slicing

Indexing accesses a single element. Slicing extracts a portion of the list.

### Syntax
```python
list_name[start:stop:step]
```
- `start`: starting index (included)
- `stop`: ending index (excluded)
- `step`: increment between indices

Negative indices count from the end of the list.


````python
# Create a list of city names
cities = ["Delhi", "Mumbai", "Chennai", "Kolkata", "Bengaluru", "Hyderabad"]
# Access the first and last elements
print("First city:", cities[0])
print("Last city:", cities[-1])
# Extract elements from index 1 to 3 (stop is excluded)
print("Slice:", cities[1:4])
# Extract every second element
print("Every second city:", cities[::2])
# Reverse the list using slicing
print("Reverse:", cities[::-1])
````

### Expected Output
```text
First city: Delhi
Last city: Hyderabad
Slice: ['Mumbai', 'Chennai', 'Kolkata']
Every second city: ['Delhi', 'Chennai', 'Bengaluru']
Reverse: ['Hyderabad', 'Bengaluru', 'Kolkata', 'Chennai', 'Mumbai', 'Delhi']
```


## 1.2 List Comprehension

List comprehension is a concise way to create a new list by iterating over an iterable and optionally applying a condition.

### Syntax
```python
[expression for item in iterable if condition]
```


````python
# List of temperatures in Celsius
temperatures = [18, 22, 25, 30, 35, 40]
# Convert every temperature to Fahrenheit
fahrenheit = [((temp * 9/5) + 32) for temp in temperatures]
# Create a list containing only temperatures above 25 Celsius
hot_days = [temp for temp in temperatures if temp > 25]

# Create a list of squares of even numbers
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
even_squares = [num ** 2 for num in numbers if num % 2 == 0]

#print("Fahrenheit:", fahrenheit)
# print("Hot days:", hot_days)
print("Even squares:", even_squares)
````

### Expected Output
```text
Fahrenheit: [64.4, 71.6, 77.0, 86.0, 95.0, 104.0]
Hot days: [30, 35, 40]
Even squares: [4, 16, 36, 64]
```


## 1.3 Nested Lists

A nested list is a list that contains other lists. It can represent tables, matrices, or structured records.


````python
# Store student names and marks as nested lists
students = [
    ["Ananya", 85, 90, 78],
    ["Rohan", 72, 88, 91],
    ["Priya", 95, 93, 89]
]


print(students[2][1])

# Calculate each student's average
for student in students:
    # student[0] is the name; remaining elements are marks
    name = student[0]
    student_marks = student[1:]
    average = sum(student_marks) / len(student_marks)
    print(name, "Average:", round(average, 2))
````

### Expected Output
```text
95
Ananya Average: 84.33
Rohan Average: 83.67
Priya Average: 92.33
```


## 1.5 Medium-Level List Exercises

3. Find the second-largest unique number in a list.

````python
# Exercise 1: Find duplicate values
values = [10, 20, 10, 30, 40, 20, 50, 10]
duplicates = [value for value in set(values) if values.count(value) > 1]
print("Duplicates:", duplicates)

# Exercise 2: Remove duplicates while preserving order
unique_ordered = []
for value in values:
    # Add the value only if it has not already been added
    if value not in unique_ordered:
        unique_ordered.append(value)
print("Unique in order:", unique_ordered)

# Exercise 3: Find second-largest unique number
unique_values = sorted(set(values))
second_largest = unique_values[-2]
print("Second largest:", second_largest)

# Exercise 4: Marks above average
marks = [65, 75, 80, 90, 55, 95]
average = sum(marks) / len(marks)
above_average = [mark for mark in marks if mark > average]
print("Average:", average)
print("Above average:", above_average)

# Exercise 5: Flatten a nested list
nested = [[1, 2], [3, 4, 5], [6], [7, 8]]
flattened = [item for sublist in nested for item in sublist]
print("Flattened:", flattened)
````

# 2. Tuples

## Introduction
A **tuple** is an ordered collection of elements that is immutable after creation.

### Key Characteristics
- Ordered.
- Immutable: cannot add, remove, or replace tuple elements directly.
- Allows duplicates.
- Supports indexing and slicing.
- Can contain different data types.
- Useful for fixed data and records.

### Tuple Syntax
```python
my_tuple = (10, 20, 30)
```

A tuple with one element requires a comma:
```python
single_item = (10,)
```


````python
# Create a tuple of fixed employee details
employee = (101, "Ananya", "Data Analyst", 55000)

# Access elements using indexing
print("Employee ID:", employee[0])
print("Name:", employee[1])
# Access a slice
print("Details:", employee[1:])
# Count how many times a value occurs
numbers = (10, 20, 10, 30, 10)
print("Count of 10:", numbers.count(10))
# Find the index of a value
print("Index of 30:", numbers.index(30))
````

### Expected Output
```text
Employee ID: 101
Name: Ananya
Details: ('Ananya', 'Data Analyst', 55000)
Count of 10: 3
Index of 30: 3
```


## 2.1 Tuple Unpacking

Tuple unpacking assigns tuple elements to individual variables. The number of variables must normally match the number of elements.


````python
# Store coordinates in a tuple
coordinates = (12.5, 18.75)

# Unpack tuple values into variables
x, y = coordinates

print("X:", x)
print("Y:", y)

# Unpack employee information
employee = (101, "Rohan", 60000)
emp_id, name, salary = employee

print(emp_id)
print(name)
print(salary)

# Use * to collect remaining values
numbers = (1, 2, 3, 4, 5)
first, *middle, last = numbers

print("First:", first)
print("Middle:", middle)
print("Last:", last)
````

### Expected Output
```text
X: 12.5
Y: 18.75
101
Rohan
60000
First: 1
Middle: [2, 3, 4]
Last: 5
```


## 2.2 Tuple and List Conversion

Although tuples are immutable, we can convert a tuple to a list, modify the list, and convert it back to a tuple.


````python
# Create a tuple
subjects = ("Python", "SQL", "Statistics")
# Convert tuple into a list
subject_list = list(subjects)
# Modify the list
subject_list.append("Machine Learning")

# Convert the modified list back to a tuple
updated_subjects = tuple(subject_list)

print("Original tuple:", subjects)
print("Updated tuple:", updated_subjects)
````

### Expected Output
```text
Original tuple: ('Python', 'SQL', 'Statistics')
Updated tuple: ('Python', 'SQL', 'Statistics', 'Machine Learning')
```

# 3. Sets

## Introduction
A **set** is an unordered collection of unique elements.

### Key Characteristics
- Does not allow duplicate elements.
- Unordered: do not rely on index positions.
- Mutable set: elements can be added or removed.
- Elements must be hashable/immutable.
- Useful for removing duplicates and performing mathematical set operations.

### Creating Sets
```python
my_set = {10, 20, 30}
empty_set = set()  # {} creates an empty dictionary, not a set
```

### Common Methods
| Method | Description |
|---|---|
| `add()` | Adds one element |
| `update()` | Adds multiple elements |
| `remove()` | Removes an element; raises error if missing |
| `discard()` | Removes an element if present |
| `pop()` | Removes and returns an arbitrary element |
| `clear()` | Removes all elements |


````python
# Create a set with duplicate values
numbers = {10, 20, 30, 20, 40, 10}
# Duplicates are automatically removed
print("Set:", numbers)
# Add one element
numbers.add(50)
print("After add:", numbers)
# Add multiple elements
numbers.update([60, 70, 80])
print("After update:", numbers)
# Remove an element if it exists
numbers.discard(100)  # No error if 100 is not present
numbers.discard(20)
print("After discard:", numbers)
````

### Expected Output
The exact display order of set elements can vary. The set contains:
```text
{10, 30, 40, 50, 60, 70, 80}
```


## 3.1 Mathematical Set Operations

Python supports common mathematical set operations:
- Union (`|`): elements from either set.
- Intersection (`&`): elements common to both.
- Difference (`-`): elements in the first set but not the second.
- Symmetric difference (`^`): elements in either set but not both.


````python
# Define two sets of course participants
python_students = {"Ananya", "Rohan", "Priya", "Aman"}
sql_students = {"Rohan", "Priya", "Neha", "Karan"}

# Union: participants in either course
print("Union:", python_students | sql_students)

# Intersection: participants in both courses
print("Intersection:", python_students & sql_students)

# Difference: Python students who are not in SQL
print("Python only:", python_students - sql_students)

# Symmetric difference: participants in exactly one course
print("Exactly one course:", python_students ^ sql_students)
````

### Expected Results
Set order may vary, but the contents are:

- **Union:** Ananya, Rohan, Priya, Aman, Neha, Karan
- **Intersection:** Rohan, Priya
- **Python only:** Ananya, Aman
- **Exactly one course:** Ananya, Aman, Neha, Karan


## 3.2 Set Membership and Subset Checks

Sets can efficiently check membership and relationships between collections.


````python
# Define sets
team_a = {"A", "B", "C"}
team_b = {"A", "B", "C", "D", "E"}
team_c = {"X", "Y"}

# Check membership
print("B is in team_a:", "B" in team_a)

# Check whether team_a is a subset of team_b
print("A subset of B:", team_a.issubset(team_b))

# Check whether team_b is a superset of team_a
print("B superset of A:", team_b.issuperset(team_a))

# Check whether two sets are disjoint
print("A and C disjoint:", team_a.isdisjoint(team_c))
````

### Expected Output
```text
B is in team_a: True
A subset of B: True
B superset of A: True
A and C disjoint: True
```


## 3.3 Remove Duplicates from a List Using a Set

Sets are commonly used to remove duplicates. However, a set does not preserve the original list order as a general guarantee. If order matters, use a different approach such as `dict.fromkeys()`.


````python
# List containing duplicate values
items = [5, 2, 5, 3, 2, 8, 3, 9]

# Remove duplicates using a set
unique_items = list(set(items))

# Remove duplicates while preserving original order
unique_order = list(dict.fromkeys(items))

print("Set-based unique values:", unique_items)
print("Order-preserving unique values:", unique_order)
````

### Expected Output
The set-based list may appear in a different order. The order-preserving result is:
```text
Order-preserving unique values: [5, 2, 3, 8, 9]
```

# 4. Dictionaries

## Introduction
A **dictionary** stores data as key-value pairs.

### Key Characteristics
- Mutable.
- Keys must be unique.
- Keys must be hashable.
- Values can be any data type.
- In modern Python, dictionaries preserve insertion order.
- Useful for representing structured records and mapping one value to another.

### Syntax
```python
student = {
    "name": "Ananya",
    "age": 21,
    "marks": 88
}
```

### Common Operations
| Operation | Description |
|---|---|
| `dict[key]` | Access value; raises KeyError if key missing |
| `get()` | Access value safely |
| `keys()` | View of keys |
| `values()` | View of values |
| `items()` | View of key-value pairs |
| `update()` | Add/update values |
| `pop()` | Remove a key and return its value |
| `setdefault()` | Return value or insert default |
| `del` | Delete a key |


````python
# Create a student dictionary
student = {
    "name": "Ananya",
    "age": 21,
    "course": "Data Analytics",
    "marks": 88
}
# Access values using keys
print("Name:", student["name"])
print("Marks:", student.get("marks"))
# Safely access a key that may not exist
print("City:", student.get("city", "Not available"))
# Update an existing value
student["marks"] = 92
# Add a new key-value pair
student["city"] = "Delhi"
# Update multiple values
student.update({"age": 22, "status": "Active"})
# Display updated dictionary
print(student)
````

### Expected Output
```text
Name: Ananya
Marks: 88
City: Not available
{'name': 'Ananya', 'age': 22, 'course': 'Data Analytics', 'marks': 92, 'city': 'Delhi', 'status': 'Active'}
```


## 4.1 Dictionary Methods

Let's explore methods for accessing, updating, and removing dictionary entries.


````python
# Create a dictionary
inventory = {"Laptop": 10, "Mouse": 25, "Keyboard": 15}

# Display all keys
print("Keys:", list(inventory.keys()))

# Display all values
print("Values:", list(inventory.values()))

# Display key-value pairs
print("Items:", list(inventory.items()))

# Safely retrieve an item
print("Laptop stock:", inventory.get("Laptop"))

# Add an item only if the key does not exist
inventory.setdefault("Monitor", 5)

# Remove and return an item
removed_stock = inventory.pop("Mouse")

# Display removed value and updated inventory
print("Removed Mouse stock:", removed_stock)
print("Updated inventory:", inventory)
````

### Expected Output
```text
Keys: ['Laptop', 'Mouse', 'Keyboard']
Values: [10, 25, 15]
Items: [('Laptop', 10), ('Mouse', 25), ('Keyboard', 15)]
Laptop stock: 10
Removed Mouse stock: 25
Updated inventory: {'Laptop': 10, 'Keyboard': 15, 'Monitor': 5}
```


## 4.2 Iterating Through a Dictionary

Use `.items()` to access keys and values together.


````python
# Dictionary containing employee salaries
salaries = {
    "Aman": 45000,
    "Neha": 60000,
    "Ravi": 50000,
    "Pooja": 40000
}

# Iterate over key-value pairs
for employee, salary in salaries.items():
    # Display each employee and salary
    print(employee, ":", salary)

# Find employees earning more than 45,000
high_salary = {
    employee: salary
    for employee, salary in salaries.items()
    if salary > 45000
}

print("High salary:", high_salary)
````

### Expected Output
```text
Aman : 45000
Neha : 60000
Ravi : 50000
Pooja : 40000
High salary: {'Neha': 60000, 'Ravi': 50000}
```


## 4.3 Dictionary Comprehension

Dictionary comprehension creates dictionaries using a concise expression.


````python
# List of numbers
numbers = [1, 2, 3, 4, 5]

# Create a dictionary mapping each number to its square
squares = {number: number ** 2 for number in numbers}

# Create a dictionary of even numbers and their cubes
even_cubes = {number: number ** 3 for number in numbers if number % 2 == 0}

print("Squares:", squares)
print("Even cubes:", even_cubes)
````

### Expected Output
```text
Squares: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
Even cubes: {2: 8, 4: 64}
```


## 4.4 Nested Dictionaries

A dictionary can contain other dictionaries. This is useful for representing structured data such as students, employees, products, or departments.


````python
# Nested dictionary representing student records
students = {
    101: {"name": "Ananya", "marks": 88, "city": "Delhi"},
    102: {"name": "Rohan", "marks": 76, "city": "Mumbai"},
    103: {"name": "Priya", "marks": 95, "city": "Chennai"}
}

# Access a nested value
print("Student 102 name:", students[102]["name"])

# Iterate through all records
for roll_no, details in students.items():
    # Extract nested information
    name = details["name"]
    marks = details["marks"]

    # Display student information
    print(roll_no, name, marks)

# Add a new student record
students[104] = {"name": "Aman", "marks": 82, "city": "Kolkata"}

print("New student:", students[104])
````

### Expected Output
```text
Student 102 name: Rohan
101 Ananya 88
102 Rohan 76
103 Priya 95
New student: {'name': 'Aman', 'marks': 82, 'city': 'Kolkata'}
```


## 4.5 Medium-Level Dictionary Exercises

1. Count the frequency of each word in a sentence.
2. Find the employee with the highest salary.
3. Group a list of students by their department.
4. Merge two dictionaries and add values when keys overlap.
5. Create a dictionary of student names and marks, then generate a new dictionary containing only students who scored at least 75.


### Solutions


````python
# Exercise 1: Word frequency
sentence = "python data python analytics data python"
word_counts = {}

# Count each word
for word in sentence.split():
    # Increase existing count or start at 1
    word_counts[word] = word_counts.get(word, 0) + 1

print("Word counts:", word_counts)

# Exercise 2: Find employee with highest salary
employees = {"Aman": 45000, "Neha": 60000, "Ravi": 50000, "Pooja": 40000}
highest_paid = max(employees, key=employees.get)
print("Highest paid:", highest_paid, employees[highest_paid])

# Exercise 3: Group students by department
student_records = [
    ("Ananya", "IT"),
    ("Rohan", "HR"),
    ("Priya", "IT"),
    ("Aman", "HR"),
    ("Neha", "Finance")
]

department_groups = {}
for name, department in student_records:
    # Create a list for the department if it is not present
    department_groups.setdefault(department, []).append(name)

print("Department groups:", department_groups)

# Exercise 4: Merge dictionaries and add overlapping values
sales_january = {"Delhi": 10000, "Mumbai": 15000, "Chennai": 12000}
sales_february = {"Delhi": 12000, "Mumbai": 10000, "Kolkata": 8000}

combined_sales = sales_january.copy()
for city, amount in sales_february.items():
    # Add to existing city sales or create a new city entry
    combined_sales[city] = combined_sales.get(city, 0) + amount

print("Combined sales:", combined_sales)

# Exercise 5: Filter students scoring at least 75
marks = {"Ananya": 88, "Rohan": 65, "Priya": 95, "Aman": 72, "Neha": 80}
qualified = {name: score for name, score in marks.items() if score >= 75}
print("Qualified:", qualified)
````

# 5. Choosing the Right Collection

| Collection | Ordered | Mutable | Duplicates | Typical Use |
|---|---|---|---|---|
| List | Yes | Yes | Allowed | Dynamic sequences |
| Tuple | Yes | No | Allowed | Fixed records or values |
| Set | No | Yes | Not allowed | Unique values and set operations |
| Dictionary | Insertion ordered | Yes | Keys unique | Key-value mapping |

### Quick Decision Guide
- Use a **list** when order matters and you need to modify the collection.
- Use a **tuple** when the data should remain fixed.
- Use a **set** when you need unique elements or set operations.
- Use a **dictionary** when you need to map keys to values.


# 6. Combined Medium-Level Challenge

## Student Result Processing

Given a list of student records, perform the following:

1. Store student records in a list of dictionaries.
2. Calculate each student's average marks.
3. Create a dictionary mapping student names to averages.
4. Use a set to identify unique subjects.
5. Create a tuple containing the top student's name and average.
6. Display students whose average is at least 80.
7. Sort students by average marks in descending order.

Try solving this independently before looking at the solution.


## Challenge Solution


````python
# Student records stored as a list of dictionaries
students = [
    {"name": "Ananya", "marks": {"Python": 85, "SQL": 90, "Statistics": 80}},
    {"name": "Rohan", "marks": {"Python": 75, "SQL": 82, "Statistics": 78}},
    {"name": "Priya", "marks": {"Python": 95, "SQL": 92, "Statistics": 96}},
    {"name": "Aman", "marks": {"Python": 80, "SQL": 70, "Statistics": 85}}
]

# Create a dictionary to store student averages
averages = {}

# Create a set to collect unique subject names
subjects = set()

# Process each student
for student in students:
    # Extract the student's name
    name = student["name"]

    # Extract the marks dictionary
    marks = student["marks"]

    # Collect subject names
    subjects.update(marks.keys())

    # Calculate the average mark
    average = sum(marks.values()) / len(marks)

    # Store the average using the student's name as the key
    averages[name] = average

# Display student averages
print("Averages:", averages)

# Display unique subjects
print("Subjects:", subjects)

# Find the student with the highest average
top_student = max(averages, key=averages.get)

# Store top student details in a tuple
top_result = (top_student, averages[top_student])
print("Top student:", top_result)

# Filter students with average >= 80
qualified = {name: avg for name, avg in averages.items() if avg >= 80}
print("Qualified:", qualified)

# Sort students by average descending
sorted_students = sorted(averages.items(), key=lambda item: item[1], reverse=True)
print("Ranking:")
for name, avg in sorted_students:
    print(name, round(avg, 2))
````

Subjects: {'Python', 'SQL', 'Statistics'}

Qualified: {'Ananya': 85.0, 'Priya': 94.33333333333333}
Ranking:
