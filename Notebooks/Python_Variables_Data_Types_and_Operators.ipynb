## 2. Python Variables

A **variable** is a name that refers to an object in memory. Variables are used to store and work with data.

Python does not require an explicit declaration of a variable's data type. The type is determined from the value assigned to it.

### 2.1 Creating variables

````python
name = "Ananya"
age = 22
salary = 35000.50
is_trainer = True

print(name)
print(age)
print(salary)
print(is_trainer)
````

**Explanation**
- `name` refers to a string object.
- `age` refers to an integer object.
- `salary` refers to a floating-point object.
- `is_trainer` refers to a Boolean object.
- The `=` symbol is the assignment operator. It assigns the object on the right to the name on the left.

**Expected output**
```text
Ananya
22
35000.5
True
```

### 2.2 Dynamic typing

A variable can be reassigned to an object of a different type. Python determines the type at runtime.

````python
value = 100
print(value, type(value))

value = "Python"
print(value, type(value))
````

**Explanation**
- Initially, `value` refers to an integer.
- After reassignment, it refers to a string.
- `type()` returns the type of the object currently referenced by the variable.

**Expected output**
```text
100 <class 'int'>
Python <class 'str'>
```

### 2.3 Multiple assignment

````python
# Assign different values to multiple variables
student_name, marks, passed = "Riya", 86, True
print(student_name, marks, passed)

# Assign the same value to multiple variables
x = y = z = 10
print(x, y, z)

# Unpack values from a sequence
first, second, third = [100, 200, 300]
print(first, second, third)
````

**Explanation**
- Multiple variables can be assigned in one statement.
- Chained assignment assigns the same object to several names.
- Unpacking assigns sequence elements to individual variables. The number of targets must match the number of values.

**Expected output**
```text
Riya 86 True
10 10 10
100 200 300
```

### 2.4 Rules for naming variables

Valid variable names:
- Can contain letters, digits and underscores.
- Must not start with a digit.
- Are case-sensitive (`score` and `Score` are different names).
- Must not be Python keywords such as `if`, `for` or `class`.
- Should use meaningful names, such as `total_marks` rather than `tm`.

Examples:

````python
student_name = "Aman"
_marks = 75
total2 = 150
studentAge = 20

print(student_name, _marks, total2, studentAge)
````

**Expected output**
```text
Aman 75 150 20
```

The following are invalid variable names and should not be run as code:
- `2marks` — starts with a digit.
- `student-name` — contains a hyphen.
- `class` — reserved Python keyword.

## 3. Python Data Types

A data type describes the kind of value an object represents and the operations that can be performed on it.

Common built-in data types include:

| Category | Data type | Example |
|---|---|---|
| Numeric | `int` | `25` |
| Numeric | `float` | `25.5` |
| Numeric | `complex` | `2 + 3j` |
| Text | `str` | `"Python"` |
| Boolean | `bool` | `True` |
| Sequence | `list` | `[10, 20]` |
| Sequence | `tuple` | `(10, 20)` |
| Set | `set` | `{10, 20}` |
| Mapping | `dict` | `{"name": "Asha"}` |
| None type | `NoneType` | `None` |

Use `type()` to inspect an object's type.

````python
values = [25, 25.5, 2 + 3j, "Python", True, [1, 2], (1, 2),
          {1, 2}, {"name": "Asha"}, None]

for value in values:
    print(value, "->", type(value))
````

**Explanation**
- `type(value)` reports the object's type.
- The `for` loop visits each value in the list and prints its value and type.
- Notice that `bool` is a subclass of `int` in Python, even though Boolean values are normally treated as a separate logical type.

**Expected output**
```text
25 -> <class 'int'>
25.5 -> <class 'float'>
(2+3j) -> <class 'complex'>
Python -> <class 'str'>
True -> <class 'bool'>
[1, 2] -> <class 'list'>
(1, 2) -> <class 'tuple'>
{1, 2} -> <class 'set'>
{'name': 'Asha'} -> <class 'dict'>
None -> <class 'NoneType'>
```

## 4. Integer (`int`)

An integer is a whole number without a decimal component. It can be positive, negative or zero. Python integers can grow to very large values, subject to available memory.

### 4.1 Creating integers

````python
a = 100
b = -45
c = 0

print(a, type(a))
print(b, type(b))
print(c, type(c))
````

**Explanation**
- `100`, `-45` and `0` are all integer values.
- Their type is `int`.

**Expected output**
```text
100 <class 'int'>
-45 <class 'int'>
0 <class 'int'>
```

### 4.2 Integer number systems

Python supports integer literals in four commonly used number systems:

| Number system | Base | Prefix | Example |
|---|---:|---|---|
| Decimal | 10 | No prefix | `25` |
| Binary | 2 | `0b` | `0b11001` |
| Octal | 8 | `0o` | `0o31` |
| Hexadecimal | 16 | `0x` | `0x19` |

All four examples below represent the decimal value 25.

````python
decimal_value = 25
binary_value = 0b11001
octal_value = 0o31
hex_value = 0x19

print(decimal_value)
print(binary_value)
print(octal_value)
print(hex_value)
````

**Explanation**
- `0b` indicates a binary literal.
- `0o` indicates an octal literal.
- `0x` indicates a hexadecimal literal.
- Python evaluates each literal as an integer.

**Expected output**
```text
25
25
25
25
```

### 4.3 Converting decimal integers to other bases

The built-in functions `bin()`, `oct()` and `hex()` return string representations of an integer in binary, octal and hexadecimal.

````python
number = 25

print(bin(number))
print(oct(number))
print(hex(number))
````

**Explanation**
- `bin(25)` returns the binary string.
- `oct(25)` returns the octal string.
- `hex(25)` returns the hexadecimal string.
- The prefixes remain in the returned strings.

**Expected output**
```text
0b11001
0o31
0x19
```

### 4.4 Converting other bases to decimal

Use `int(value, base)` to convert a string written in a specified base into a decimal integer.

````python
print(int("11001", 2))
print(int("31", 8))
print(int("19", 16))
````

**Explanation**
- `int("11001", 2)` interprets the string as binary.
- `int("31", 8)` interprets it as octal.
- `int("19", 16)` interprets it as hexadecimal.

**Expected output**
```text
25
25
25
```

## 5. Floating-Point (`float`)

A float represents a real number with a fractional or decimal component. Examples include `10.5`, `-3.14` and `0.0`.

### 5.1 Creating float values

````python
price = 99.99
temperature = -2.5
percentage = 85.0

print(price, type(price))
print(temperature, type(temperature))
print(percentage, type(percentage))
````

**Explanation**
- A number written with a decimal point is normally a `float`.
- `85.0` is a float even though it has no non-zero fractional part.

**Expected output**
```text
99.99 <class 'float'>
-2.5 <class 'float'>
85.0 <class 'float'>
```

### 5.2 Scientific notation

Scientific notation is useful for representing very large or very small numbers. Python uses `e` or `E` to indicate a power of ten.

````python
large_number = 2.5e6
small_number = 3.2e-4

print(large_number)
print(small_number)
````

**Explanation**
- `2.5e6` means \(2.5 \times 10^6\).
- `3.2e-4` means \(3.2 \times 10^{-4}\).

**Expected output**
```text
2500000.0
0.00032
```

### 5.3 Float arithmetic and conversion

````python
a = 10.5
b = 2

print(a + b)
print(a * b)
print(float(25))
print(int(25.9))
````

**Explanation**
- Arithmetic involving a float generally produces a float.
- `float(25)` converts an integer to a float.
- `int(25.9)` truncates the fractional part; it does not round to the nearest integer.

**Expected output**
```text
12.5
21.0
25.0
25
```

### 5.4 Floating-point precision

Some decimal fractions cannot be represented exactly in binary floating-point. This can produce results that look surprising.

````python
result = 0.1 + 0.2
print(result)
print(round(result, 2))
````

**Explanation**
- The internal binary approximations of `0.1` and `0.2` are not exact.
- Therefore, the direct sum may display as `0.30000000000000004`.
- `round(result, 2)` rounds the displayed numeric result to two decimal places.
- For decimal arithmetic requiring exact decimal representation, consider Python's `decimal.Decimal` class.

**Expected output**
```text
0.30000000000000004
0.3
```

### 5.5 Decimal module for decimal arithmetic

`Decimal` can represent decimal values exactly when created from strings.

````python
from decimal import Decimal

a = Decimal("0.1")
b = Decimal("0.2")

print(a + b)
````

**Explanation**
- Creating `Decimal` from strings avoids importing the binary approximation of a float.
- This is useful in financial calculations where decimal precision matters.

**Expected output**
```text
0.3
```

## 6. Complex Numbers (`complex`)

A complex number contains a real part and an imaginary part. Python uses `j` to represent the imaginary unit.

A complex number has the form:

\[
a + bj
\]

where `a` is the real part and `b` is the imaginary coefficient.

### 6.1 Creating complex numbers

````python
z = 4 + 3j

print(z)
print(type(z))
print(z.real)
print(z.imag)
````

**Explanation**
- `4` is the real part.
- `3` is the imaginary coefficient.
- `.real` returns the real component.
- `.imag` returns the imaginary component.

**Expected output**
```text
(4+3j)
<class 'complex'>
4.0
3.0
```

### 6.2 Creating complex numbers with `complex()`

````python
z1 = complex(2, 5)
z2 = complex(7, -1)

print(z1)
print(z2)
````

**Explanation**
- `complex(real, imag)` constructs a complex number.
- `complex(2, 5)` creates `2 + 5j`.
- `complex(7, -1)` creates `7 - 1j`.

**Expected output**
```text
(2+5j)
(7-1j)
```

### 6.3 Complex number operations

````python
a = 2 + 3j
b = 1 + 4j

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Conjugate:", a.conjugate())
print("Modulus:", abs(a))
````

**Explanation**
- Complex numbers support addition, subtraction, multiplication and division.
- The conjugate changes the sign of the imaginary component.
- `abs()` returns the magnitude: for \(a+bj\), it is \(\sqrt{a^2+b^2}\).

**Expected output**
```text
Addition: (3+7j)
Subtraction: (1-1j)
Multiplication: (-10+11j)
Division: (0.8235294117647058-0.29411764705882354j)
Conjugate: (2-3j)
Modulus: 3.605551275463989
```

### 7.1 Creating strings

**Explanation**

- Triple quotes can preserve line breaks in multiline strings.
- The opening and closing quotation marks must match.

**Expected output**

````python
message1 = "It's a Python class."
message2 = 'She said, "Hello!"'

print(message1)
print(message2)
````

### 7.3 Escape characters

Escape sequences begin with a backslash (`\\`) and represent special characters.

| Escape sequence | Meaning |
|---|---|
| `\n` | New line |
| `\t` | Horizontal tab |
| `\\` | Backslash |
| `\'` | Single quote |
| `\"` | Double quote |

````python
print("Hello\nPython")
print("Name:\tAsha")
print("C:\\Users\\Student")
print("It\'s easy!")
````

**Explanation**
- `\n` moves the text to a new line.
- `\t` inserts a tab.

- `\'` prints a single quote.

**Expected output**
```text
Hello

Name:   Asha
C:\Users\Student

```

### 7.4 Raw strings

A raw string treats backslashes as literal characters rather than interpreting them as escape sequences. It is useful for Windows paths and regular expressions.

````python
path = r"C:\Users\Student\Documents"
print(path)
````

**Explanation**
- The `r` prefix creates a raw string.
- Backslashes are preserved as written.

**Expected output**
```text
C:\Users\Student\Documents
```

### 7.5 String indexing and slicing

Each character in a string has an index. Positive indexing starts at `0`; negative indexing starts at `-1` from the end.

````python
language = "PYTHON"

print(language[0])
print(language[2])
print(language[-1])
print(language[0:3])
print(language[2:])
print(language[:4])
print(language[::-1])
````

**Explanation**
- `language[0]` returns the first character.
- `language[-1]` returns the last character.
- `language[0:3]` returns characters at indices 0, 1 and 2; the stop index is excluded.
- `language[::-1]` returns the string in reverse order.

**Expected output**
```text
P
T
N
PYT
THON
PYTH
NOHTYP
```

- `upper()` and `lower()` change letter case.
- `title()` capitalizes the first letter of each word.
- `strip()` removes leading and trailing whitespace.
- `replace()` replaces matching text.
- `count()` counts occurrences.
- `find()` returns the starting index, or `-1` if not found.
- `startswith()` and `endswith()` return Boolean results.

**Expected output**
```text

  python programming  
  Python Programming
python programming
  Java programming  
2
9
True
True
```

Boolean values are commonly used in comparisons and conditional statements.

````python
is_valid = True
is_logged_in = False

print(is_valid, type(is_valid))
print(is_logged_in, type(is_logged_in))
````

**Explanation**

## 9. F-Strings (Formatted String Literals)

An f-string makes it easy to insert variable values and expressions into a string. Place `f` or `F` before the opening quotation mark and put expressions inside curly braces `{}`.

````python
name = "Riya"
age = 22
marks = 87.456

print(f"My name is {name}.")
print(f"I am {age} years old.")
print(f"My marks are {marks:.2f}.")
print(f"Next year, I will be {age + 1}.")
````

**Explanation**
- `{name}` inserts the value of `name`.
- `{age}` inserts the value of `age`.
- `{marks:.2f}` formats the number to two digits after the decimal point.
- `{age + 1}` evaluates an expression inside the f-string.

**Expected output**
```text
My name is Riya.
I am 22 years old.
My marks are 87.46.
Next year, I will be 23.
```

### 9.1 F-string with a variable equal to zero

````python
x = 0
print(f"x stores {x}")

x = 10
print(f"x stores {x}")
````

**Explanation**
- The text `x stores` is displayed as written.
- `{x}` is replaced by the current value of the variable.
- The same f-string pattern works when the value changes.

**Expected output**
```text
x stores 0
x stores 10
```

## 10. Type Conversion

Type conversion means converting a value from one data type to another.

Common conversion functions:
- `int()` — converts a compatible value to an integer.
- `float()` — converts a compatible value to a float.
- `str()` — converts a value to a string.
- `bool()` — converts a value to a Boolean.
- `complex()` — creates a complex number.

````python
number_text = "250"
price_text = "99.50"

number = int(number_text)
price = float(price_text)
message = str(number)
status = bool(number)

print(number, type(number))
print(price, type(price))
print(message, type(message))
print(status, type(status))
````

**Explanation**
- Numeric strings can be converted using `int()` or `float()`.
- `str()` converts a number into text.
- A non-zero integer converts to `True` with `bool()`.

**Expected output**
```text
250 <class 'int'>
99.5 <class 'float'>
250 <class 'str'>
True <class 'bool'>
```

**Important:** `int("hello")` raises a `ValueError` because `"hello"` is not a valid integer representation.

## 12. Object Identity: `id()` Function

The built-in `id()` function returns an integer that identifies an object during its lifetime. In CPython, this value is typically related to the object's memory address. Python's language specification does not require it to be a physical memory address.

**Syntax**
```python
id(object)
```

````python
x = 10

print(x)
print(id(x))
````

**Explanation**
- `x` refers to an integer object.
- `print(x)` displays its value.
- `id(x)` returns the object's identity.
- The actual identity number varies between runs and Python implementations.

**Expected output**
```text
10
An integer identity value, which varies by environment
```

### 12.1 Two variables referring to the same object

````python
a = [10, 20, 30]
b = a

print("ID of a:", id(a))
print("ID of b:", id(b))
print("Same object:", a is b)
````

**Explanation**
- `a` creates a list object.
- `b = a` makes `b` refer to that same list object.
- Therefore, `id(a)` and `id(b)` are equal.
- `a is b` evaluates to `True`.

**Expected output**
```text
ID of a: <same integer identity>
ID of b: <same integer identity>
Same object: True
```

### 12.2 Two variables with equal values but different objects

````python
a = [10, 20, 30]
b = [10, 20, 30]

print("Equal values:", a == b)
print("Same object:", a is b)
print("IDs are equal:", id(a) == id(b))
````

**Explanation**
- The two list objects contain the same elements, so `a == b` is `True`.
- They are separate list objects, so `a is b` is `False`.
- Their identities are therefore different while both objects are alive.

**Expected output**
```text
Equal values: True
Same object: False
IDs are equal: False
```

## 13. Python Operators

Operators are symbols or keywords used to perform operations on values and variables.

Python provides several categories of operators:

1. Arithmetic operators
2. Assignment operators
3. Comparison (relational) operators
4. Logical operators
5. Bitwise operators
6. Membership operators
7. Identity operators

### 13.1 Arithmetic operators

Arithmetic operators perform mathematical calculations.

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `10 + 3` |
| `-` | Subtraction | `10 - 3` |
| `*` | Multiplication | `10 * 3` |
| `/` | Division | `10 / 3` |
| `//` | Floor division | `10 // 3` |
| `%` | Modulus (remainder) | `10 % 3` |
| `**` | Exponentiation | `10 ** 3` |

````python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)
````

**Explanation**
- `/` performs true division and returns a float.
- `//` performs floor division, rounding the quotient down.
- `%` returns the remainder.
- `**` raises a number to a power.

**Expected output**
```text
Addition: 13
Subtraction: 7
Multiplication: 30
Division: 3.3333333333333335
Floor division: 3
Modulus: 1
Exponentiation: 1000
```

### 13.2 Assignment operators

Assignment operators assign values to variables and can update their values.

````python
x = 10
print("Initial:", x)

x += 5
print("After += 5:", x)

x -= 3
print("After -= 3:", x)

x *= 2
print("After *= 2:", x)

x /= 4
print("After /= 4:", x)

x = 10
x //= 3
print("After //= 3:", x)

x = 10
x %= 3
print("After %= 3:", x)

x = 2
x **= 3
print("After **= 3:", x)
````

**Explanation**
- `x += 5` is equivalent to `x = x + 5`.
- `x -= 3` is equivalent to `x = x - 3`.
- Similar shorthand forms exist for multiplication, division, floor division, modulus and exponentiation.
- The example resets `x` before some operations so each result is easy to understand.

**Expected output**
```text
Initial: 10
After += 5: 15
After -= 3: 12
After *= 2: 24
After /= 4: 6.0
After //= 3: 3
After %= 3: 1
After **= 3: 8
```

### 13.3 Comparison operators

Comparison operators compare two values and return `True` or `False`.

````python
a = 10
b = 5

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= 10)
print(b <= 5)
````

**Explanation**
- `==` checks equality; `=` performs assignment.
- `!=` checks inequality.
- `>`, `<`, `>=` and `<=` compare magnitudes.

**Expected output**
```text
False
True
True
False
True
True
```

### 13.4 Logical operators

Logical operators combine or reverse Boolean conditions.

| Operator | Meaning |
|---|---|
| `and` | True if both conditions are true |
| `or` | True if at least one condition is true |
| `not` | Reverses the truth value |

````python
age = 25
has_id = True

print(age >= 18 and has_id)
print(age < 18 or has_id)
print(not has_id)
print(True and False)
print(True or False)
print(not True)
````

**Explanation**
- `and` requires both conditions to be truthy.
- `or` requires at least one condition to be truthy.
- `not` reverses a Boolean result.
- `and` and `or` use short-circuit evaluation: Python may skip evaluating the right-hand expression when the result is already determined.

**Expected output**
```text
True
True
False
False
True
False
```

### 13.5 Bitwise operators

Bitwise operators work on the binary representation of integers.

| Operator | Meaning |
|---|---|
| `&` | Bitwise AND |
| `|` | Bitwise OR |
| `^` | Bitwise XOR |
| `~` | Bitwise NOT |
| `<<` | Left shift |
| `>>` | Right shift |

````python
a = 5   # Binary: 0101
b = 3   # Binary: 0011

print("AND:", a & b)
print("OR:", a | b)
print("XOR:", a ^ b)
print("NOT:", ~a)
print("Left shift:", a << 1)
print("Right shift:", a >> 1)
````

**Explanation**
- `5` is `0101` and `3` is `0011` in four-bit binary notation.
- AND keeps a bit set only when both corresponding bits are set.
- OR sets a bit when either corresponding bit is set.
- XOR sets a bit when the corresponding bits differ.
- `~a` inverts the bits; Python integers use an unbounded signed representation, so `~5` evaluates to `-6`.
- Left shift by one moves bits left; right shift by one moves bits right.

**Expected output**
```text
AND: 1
OR: 7
XOR: 6
NOT: -6
Left shift: 10
Right shift: 2
```

### 13.6 Membership operators

Membership operators check whether a value exists in a sequence or collection.

- `in` returns `True` if an item is present.
- `not in` returns `True` if an item is absent.

````python
languages = ["Python", "Java", "SQL"]

print("Python" in languages)
print("C++" in languages)
print("C++" not in languages)

text = "Data Analytics"
print("Data" in text)
````

**Explanation**
- `"Python" in languages` checks whether the list contains `"Python"`.
- `"C++" not in languages` checks that `"C++"` is absent.
- For strings, membership checks whether a substring occurs in the string.

**Expected output**
```text
True
False
True
True
```

### 13.7 Identity operators

Identity operators check whether two references point to the same object.

- `is` checks whether two references identify the same object.
- `is not` checks whether they identify different objects.

````python
list_a = [1, 2, 3]
list_b = list_a
list_c = [1, 2, 3]

print(list_a is list_b)
print(list_a is list_c)
print(list_a is not list_c)
````

**Explanation**
- `list_b = list_a` makes both names refer to the same list.
- `list_c` creates a separate list with equal contents.
- `is` checks identity, not equality.

**Expected output**
```text
True
False
True
```

## 14. Operator Precedence and Associativity

When an expression contains multiple operators, Python follows precedence rules to decide which operation is evaluated first.

### 14.1 Common operator precedence (high to low)

| Priority | Operators / expressions |
|---:|---|
| 1 | Parentheses and grouping: `( )` |
| 2 | Exponentiation: `**` |
| 3 | Unary operators: `+x`, `-x`, `~x` |
| 4 | Multiplication, division, floor division, modulus: `*`, `/`, `//`, `%` |
| 5 | Addition and subtraction: `+`, `-` |
| 6 | Shifts: `<<`, `>>` |
| 7 | Bitwise AND: `&` |
| 8 | Bitwise XOR: `^` |
| 9 | Bitwise OR: `|` |
| 10 | Comparisons, membership and identity |
| 11 | Logical NOT: `not` |
| 12 | Logical AND: `and` |
| 13 | Logical OR: `or` |

This is a simplified teaching table of common operators. Parentheses are recommended whenever they make an expression clearer.

````python
result = 10 + 2 * 3
print(result)

result = (10 + 2) * 3
print(result)

result = 2 ** 3 ** 2
print(result)
````

**Explanation**
- Multiplication has higher precedence than addition, so `10 + 2 * 3` is `10 + 6`.
- Parentheses are evaluated first, so `(10 + 2) * 3` is `36`.
- Exponentiation is right-associative, so `2 ** 3 ** 2` is `2 ** (3 ** 2)`.

**Expected output**
```text
16
36
512
```

### 14.2 Associativity

Associativity determines the order of evaluation when operators have the same precedence.

- Most arithmetic operators, such as `+`, `-`, `*` and `/`, group from left to right.
- Exponentiation (`**`) groups from right to left.

````python
print(20 - 5 - 3)
print(100 / 10 * 2)
print(2 ** 3 ** 2)
````

**Explanation**
- `20 - 5 - 3` becomes `(20 - 5) - 3`, which is `12`.
- `100 / 10 * 2` becomes `(100 / 10) * 2`, which is `20.0`.
- `2 ** 3 ** 2` becomes `2 ** (3 ** 2)`, which is `512`.

**Expected output**
```text
12
20.0
512
```

### 14.3 Combining arithmetic, comparison and logical operators

````python
age = 22
marks = 85

result = age >= 18 and marks >= 40
print(result)

calculation = 10 + 5 * 2 > 18
print(calculation)
````

**Explanation**
- In the first expression, both comparisons are evaluated before `and`.
- In the second expression, multiplication happens before addition, and arithmetic happens before comparison.
- Parentheses can make complex conditions easier to read.

**Expected output**
```text
True
True
```

## 15. Integrated Example: Student Result Calculator

This example combines variables, input, type conversion, arithmetic, comparison, Boolean logic and f-strings.

````python
student_name = input("Enter student name: ")
mark1 = float(input("Enter marks in subject 1: "))
mark2 = float(input("Enter marks in subject 2: "))
mark3 = float(input("Enter marks in subject 3: "))

total = mark1 + mark2 + mark3
average = total / 3
passed = mark1 >= 40 and mark2 >= 40 and mark3 >= 40

print("\n----- Student Result -----")
print(f"Student: {student_name}")
print(f"Total marks: {total:.2f}")
print(f"Average: {average:.2f}")
print(f"Passed all subjects: {passed}")
````

**Explanation**

2. `float()` converts the entered marks into numeric values.
3. The total is calculated by adding all three marks.
4. The average is calculated by dividing the total by three.
5. The logical `and` operator checks whether every subject mark is at least 40.
6. F-strings display the results in a readable format.

**Example interaction**
