# Python File Handling — First-Year Students
## Beginner-friendly, not too advanced

**Learning objectives:** understand why files are needed, open/read/write/append text files, use `with`, read lines, and complete simple practice exercises.

Files help preserve information after a program finishes. Examples include saving student details, notes, marks, and simple records.

## 1. Opening a File

Python uses `open()`.

**Syntax**
```python
file = open("filename.txt", "mode")
```

- Filename: file to work with.
- Mode: operation such as reading or writing.
- File variable: used to perform operations.

### Common modes
| Mode | Meaning |
|---|---|
| `r` | Read an existing file |
| `w` | Write; creates or overwrites a file |
| `a` | Append at the end; creates if missing |
| `x` | Create a new file; error if it already exists |

For this lesson, focus mainly on `r`, `w`, and `a`.

## 2. Writing to a File — `w` Mode
`w` creates a file if it does not exist. If it exists, its old content is overwritten. `\n` moves to a new line.

````python
# Open student.txt in write mode
file = open("student.txt", "w")

# Write two lines into the file
file.write("Name: Ananya\n")
file.write("Course: Python Programming\n")

# Close the file after writing
file.close()
print("Data written successfully.")
````

### Expected output
```text
Data written successfully.
```

### File content
```text
Name: Ananya
Course: Python Programming
```
`write()` writes text to a file. `close()` closes the file after use.

## 3. Reading a File — `r` Mode
Read mode is used to access content in an existing file. The file must exist.

````python
# Open student.txt in read mode
file = open("student.txt", "r")

# Read the complete file content
content = file.read()

# Display the content
print(content)

# Close the file
file.close()
````

### Expected output
```text
Name: Ananya
Course: Python Programming
```

- `read()` returns the complete file content as a string.
- `r` mode does not modify the file.

## 4. Reading One Line at a Time — `readline()`
`readline()` reads one line each time it is called.

````python
# Open the file for reading
file = open("student.txt", "r")

# Read the first line
line1 = file.readline()

# Read the second line
line2 = file.readline()

# Display the lines
print("Line 1:", line1, end="")
print("Line 2:", line2, end="")

# Close the file
file.close()
````

### Expected output
```text
Line 1: Name: Ananya
Line 2: Course: Python Programming
```

## 5. Reading Lines Using a `for` Loop
We can loop over a file to process each line individually.

````python
# Open the file in read mode
file = open("student.txt", "r")

# Process each line in the file
for line in file:
    # strip() removes surrounding whitespace and newline
    print(line.strip())

# Close the file
file.close()
````

### Expected output
```text
Name: Ananya
Course: Python Programming
```

## 6. Appending to a File — `a` Mode
Append mode adds new content at the end without deleting the old content.

````python
# Open student.txt in append mode
file = open("student.txt", "a")

# Add a new line at the end
file.write("Year: First Year")

# Close the file
file.close()

print("New data appended.")
````

### Expected output
```text
New data appended.
```

Let's read the file to verify the appended content.

````python
# Open the updated file
file = open("student.txt", "r")

# Display all content
print(file.read())

# Close the file
file.close()
````

### Expected output
```text
Name: Ananya
Course: Python Programming
Year: First Year
```

## 7. The `with` Statement — Recommended Method

The `with` statement automatically closes the file after the indented block finishes. It is easier and safer than remembering to call `close()`.

**Syntax**
```python
with open("filename.txt", "mode") as file:
    # Perform file operation
```


````python
# Open the file using the with statement
with open("student.txt", "r") as file:
    # Read the content
    content = file.read()

    # Display it
    print(content)

# Python automatically closes the file after the block
print("File operation completed.")
````

### Expected output
```text
Name: Ananya
Course: Python Programming
Year: First Year

File operation completed.
```

**Teaching tip:** Use `with open(...)` in new programs whenever possible.

## 8. `read()`, `readline()`, and `readlines()`
| Method | Purpose |
|---|---|
| `read()` | Reads all content as a string |
| `readline()` | Reads one line |
| `readlines()` | Reads all lines into a list |

`readlines()` includes newline characters (`\n`) in the strings when present.

````python
# Read all lines into a list
with open("student.txt", "r") as file:
    lines = file.readlines()

# Display the list of lines
print(lines)
````

### Expected output
```text
['Name: Ananya\n', 'Course: Python Programming\n', 'Year: First Year\n']
```

## 9. Writing Multiple Lines — `writelines()`
`writelines()` writes multiple strings. It does **not** automatically add new lines, so include `\n` yourself.

````python
# Prepare a list of subjects, each ending with a newline
subjects = ["Python\n", "Mathematics\n", "Physics\n", "English\n"]

# Write all list elements to subjects.txt
with open("subjects.txt", "w") as file:
    file.writelines(subjects)

print("Subjects saved.")
````

### Expected output
```text
Subjects saved.
```

The file contains one subject per line.

## 10. Practical Example — Save Student Details
Accept student information and write it into a text file.

````python
# Get details from the user
name = input("Enter student name: ")
roll_number = input("Enter roll number: ")
department = input("Enter department: ")

# Save details into a file
with open("student_details.txt", "w") as file:
    file.write(f"Name: {name}\n")
    file.write(f"Roll Number: {roll_number}\n")
    file.write(f"Department: {department}\n")

print("Student details saved successfully.")
````

### Sample input
```text
Enter student name: Riya Sharma
Enter roll number: 101
Enter department: Computer Science
```
### Output
```text
Student details saved successfully.
```
The file will contain the entered details.

## 11. Practical Example — Count Lines and Words
Read a file and use Python's built-in string methods to count lines and words.

````python
# Read the file
with open("student.txt", "r") as file:
    content = file.read()

# Split content into lines and words
lines = content.splitlines()
words = content.split()

# Display counts
print("Line count:", len(lines))
print("Word count:", len(words))
````

### Expected output for the current `student.txt`
```text
Line count: 3
Word count: 8
```

## 12. Handling a Missing File
If a file does not exist in read mode, Python raises `FileNotFoundError`. A simple `try-except` can display a friendly message.

````python
# Try opening a file that may not exist
try:
    with open("unknown_file.txt", "r") as file:
        print(file.read())

# Handle missing-file error
except FileNotFoundError:
    print("File not found. Check the filename.")
````

### Expected output if the file is missing
```text
File not found. Check the filename.
```

# 15. Quick Revision

| Function/Method | Purpose |
|---|---|
| `open()` | Opens a file |
| `close()` | Closes a file |
| `read()` | Reads all content |
| `readline()` | Reads one line |
| `readlines()` | Reads lines into a list |
| `write()` | Writes text |
| `writelines()` | Writes multiple strings |
| `with` | Automatically closes the file |
| `r` | Read |
| `w` | Write/overwrite |
| `a` | Append |

### Remember
- Use `r` to read.
- Use `w` to write, but it overwrites existing content.
- Use `a` to append without deleting existing content.
- Use `\n` to create a new line.
- Prefer `with open(...)` because it automatically closes the file.
